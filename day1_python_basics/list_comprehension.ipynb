# List — Danh Sách

**Đặc điểm:** Mutable, ordered, indexed, cho phép phần tử trùng lặp.

## Tạo List

Dùng `[]`, có thể chứa nhiều kiểu dữ liệu khác nhau.

````python
ds = ["táo", "chuối", "cam"]
ds_hon_hop = [10, "Python", 3.14, True]
ds_rong = []
print(ds)
````

## Truy Cập — Index & Slicing

- `ds[i]` — truy cập theo index, bắt đầu từ `0`
- `ds[-1]` — index âm đếm từ cuối
- `ds[start:end]` — slicing, lấy từ `start` đến `end-1`

````python
ds = ["táo", "chuối", "cam", "xoài"]
print(ds[0])    # táo
````

````python
print(ds[-1])   # xoài
````

````python
print(ds[1:3])  # ['chuối', 'cam']
````

## Thêm Phần Tử — `append` / `insert` / `extend`

- `append(item)` — thêm 1 phần tử vào cuối
- `insert(i, item)` — chèn 1 phần tử tại vị trí `i`
- `extend(iterable)` — nối thêm nhiều phần tử từ một iterable

> `append([1,2])` thêm nguyên cả list vào như 1 phần tử; `extend([1,2])` thêm từng phần tử.

````python
ds = []
for i in range(3):
    x = input(f"Nhập phần tử {i+1}: ")
    ds.append(x)
print(ds)
````

````python
ds = ["táo", "chuối", "cam"]
ds.insert(1, "nho")
print(ds)  # ['táo', 'nho', 'chuối', 'cam']
````

````python
ds = ["táo", "chuối"]
ds.extend(["cam", "xoài"])
print(ds)  # ['táo', 'chuối', 'cam', 'xoài']
````

## Xóa Phần Tử — `remove` / `pop` / `clear` / `del`

- `remove(value)` — xóa lần xuất hiện **đầu tiên** theo giá trị
- `pop(i)` — xóa và trả về phần tử tại `i` (mặc định: cuối)
- `clear()` — xóa toàn bộ
- `del ds[i]` — xóa phần tử tại index `i`, hoặc xóa cả list với `del ds`

> `remove()` báo lỗi `ValueError` nếu giá trị không tồn tại trong list.

````python
ds = ["táo", "chuối", "táo", "cam"]
ds.remove("táo")
print(ds)  # ['chuối', 'táo', 'cam'] — chỉ xóa 'táo' đầu tiên
````

````python
ds = ["táo", "chuối", "cam"]
phan_tu = ds.pop()
print(phan_tu)  # cam
print(ds)       # ['táo', 'chuối']
````

````python
ds = ["táo", "chuối", "cam"]
ds.clear()
print(ds)  # []
````

````python
ds = ["táo", "chuối", "cam", "xoài"]
del ds[1]
print(ds)  # ['táo', 'cam', 'xoài']
````

## Tiện Ích — `sort` / `len`

- `sort()` — sắp xếp tại chỗ (in-place), dùng `reverse=True` để đảo chiều
- `len(ds)` — trả về số lượng phần tử

````python
so = [3, 1, 4, 1, 5, 9, 2]
so.sort()
print(so)               # [1, 1, 2, 3, 4, 5, 9]
so.sort(reverse=True)
print(so)               # [9, 5, 4, 3, 2, 1, 1]
````

````python
ds = ["táo", "chuối", "cam"]
print(len(ds))  # 3
````

## Tra Cứu & Đảo Ngược — `reverse` / `count` / `index`

- `reverse()` — đảo ngược thứ tự list **tại chỗ** (in-place), không trả về gì
- `count(value)` — đếm số lần xuất hiện của `value` trong list
- `index(value)` — trả về **index đầu tiên** của `value`; báo lỗi `ValueError` nếu không tìm thấy

````python
ds = ["táo", "chuối", "cam", "xoài"]
ds.reverse()
print(ds)  # ['xoài', 'cam', 'chuối', 'táo']
````

````python
ds = ["táo", "chuối", "táo", "cam", "táo"]
print(ds.count("táo"))   # 3
print(ds.count("xoài"))  # 0
````

````python
ds = ["táo", "chuối", "táo", "cam"]
print(ds.index("táo"))    # 0 — chỉ trả về index đầu tiên
print(ds.index("cam"))    # 3
# ds.index("xoài")        # ValueError: 'xoài' is not in list
````

## Biến Đổi Phần Tử — `map()`

- `map(func, iterable)` — áp dụng `func` lên **từng phần tử**, trả về một `map object`
- Bọc bằng `list()` để chuyển kết quả thành list
- Dùng `lambda` cho hàm đơn giản, dùng `def` cho hàm phức tạp

> `map()` không thay đổi list gốc, nó tạo ra một iterable mới.

````python
so = [1, 2, 3, 4, 5]

ket_qua = list(map(lambda x: x * 2, so))
print(ket_qua)  # [2, 4, 6, 8, 10]
print(so)       # [1, 2, 3, 4, 5] — list gốc không đổi
````

````python
ds = ["táo", "chuối", "cam"]

viet_hoa = list(map(str.upper, ds))
print(viet_hoa)  # ['TÁO', 'CHUỐI', 'CAM']
````

````python
def binh_phuong(x):
    return x ** 2

so = [1, 2, 3, 4, 5]
print(list(map(binh_phuong, so)))  # [1, 4, 9, 16, 25]
````

````python
R, C = map(int, input().split())

matrix = [] # Khởi tạo list rỗng để chứa ma trận (list các list)
for _ in range(R): # Lặp R lần (cho mỗi hàng)
    row_str = input().split() # Đọc dòng "1 2 3" -> ['1', '2', '3']
    row_int = list(map(int, row_str)) # Chuyển thành list số nguyên [1, 2, 3]
    matrix.append(row_int) # Thêm hàng vào ma trận

print(f"Đã đọc ma trận {R}x{C}:")
for row in matrix:
    print(row)
````

## Lọc Phần Tử — `filter()`

- `filter(func, iterable)` — giữ lại các phần tử mà `func` trả về `True`
- Cũng trả về **filter object** (lazy), bọc `list()` để lấy kết quả
- Truyền `None` thay `func` để lọc các giá trị truthy (loại bỏ `0`, `""`, `None`, `False`)

````python
so = [1, 2, 3, 4, 5, 6, 7, 8]

chan = list(filter(lambda x: x % 2 == 0, so))
print(chan)  # [2, 4, 6, 8]
````

````python
du_lieu = [0, "hello", "", None, 42, False, "Python"]

sach = list(filter(None, du_lieu))
print(sach)  # ['hello', 42, 'Python']
so = list(filter(lambda x: type(x) == int ,du_lieu))
print(so)
````

## Ghép Nhiều List — `zip()`

- `zip(a, b, ...)` — ghép các iterable theo từng cặp theo vị trí (index)
- Trả về **zip object** (lazy), bọc `list()` để lấy kết quả
- Dừng ở iterable **ngắn nhất** nếu các list có độ dài khác nhau

> Dùng `zip()` để duyệt song song nhiều list cùng lúc thay vì dùng index.

````python
ten = ["An", "Bình", "Chi", "Hieu"]
diem = [8.5, 7.0, 9.0]

ket_qua = list(zip(ten, diem))
print(ket_qua)  # [('An', 8.5), ('Bình', 7.0), ('Chi', 9.0)]
# Dùng kết quả của zip để tạo dict
data = dict(ket_qua)
print(data)
````

````python
ten = ["An", "Bình", "Chi"]
diem = [8.5, 7.0, 9.0]

for t, d in zip(ten, diem):
    print(f"{t}: {d}")
````

````python
ten = ["An", "Bình", "Chi"]
diem = [8.5, 7.0, 9.0]

# Chuyển thành dict bằng zip
bang_diem = dict(zip(ten, diem))
print(bang_diem)  # {'An': 8.5, 'Bình': 7.0, 'Chi': 9.0}
````

## Hàm Ẩn Danh — `lambda`

Cú pháp: `lambda tham_so: bieu_thuc`

- Hàm không có tên, viết gọn trong 1 dòng
- Chỉ chứa **1 biểu thức**, không dùng `return`, không có `if/else` nhiều dòng
- Thường dùng làm tham số cho `map()`, `filter()`, `sort(key=...)`

> Dùng `def` khi logic phức tạp hơn 1 dòng.

````python
# def thường vs lambda
def cong_hai(x):
    return x + 2

cong_hai_lambda = lambda x: x + 2

print(cong_hai(5))        # 7
print(cong_hai_lambda(5)) # 7
````

````python
# nhiều tham số
tong = lambda x, y: x + y
print(tong(3, 4))  # 7

# có điều kiện (ternary)
chia_het = lambda x: "chẵn" if x % 2 == 0 else "lẻ"
print(chia_het(4))  # chẵn
print(chia_het(7))  # lẻ
````

````python
# dùng lambda với sort(key=...)
hoc_sinh = [
    {"ten": "An",   "diem": 8.5},
    {"ten": "Bình", "diem": 6.0},
    {"ten": "Chi",  "diem": 9.0},
]

hoc_sinh.sort(key=lambda hs: hs["diem"], reverse=True)
for hs in hoc_sinh:
    print(f"{hs['ten']}: {hs['diem']}")
````

### Luyện tập

- **Bài 1:** Có list điểm `[5, 8, 3, 9, 6, 4, 7]`. Dùng `lambda` + `filter()` để lấy các điểm **>= 6**, sau đó dùng `lambda` + `map()` để nhân mỗi điểm với `10` (quy về thang 100).
- **Bài 2:** Có list sản phẩm sau, dùng `lambda` + `sort()` để sắp xếp theo `gia` tăng dần, nếu giá bằng nhau thì sắp xếp theo `ten` theo alphabet:
```python
san_pham = [
    {"ten": "Bút",   "gia": 5000},
    {"ten": "Vở",    "gia": 15000},
    {"ten": "Thước", "gia": 5000},
    {"ten": "Tẩy",   "gia": 3000},
]
```

````python
# Viết code của bạn ở đây
# B1:
ds = [5, 8, 3, 9, 6, 4, 7]
ds_7plus = list(filter(lambda x: x >= 6, ds))
ds_x10 = list(map(lambda x: x * 10, ds_7plus))
print(ds_x10)
````

````python
# B2:
san_pham = [
    {"ten": "Bút",   "gia": 5000},
    {"ten": "Vở",    "gia": 15000},
    {"ten": "Thước", "gia": 5000},
    {"ten": "Tẩy",   "gia": 3000},
]

san_pham.sort(key= lambda x: (x["gia"], x["ten"]), reverse=False)
print(san_pham)
````

## Chuỗi → List — `str.split()`

- `s.split(sep)` — tách chuỗi thành list dựa theo ký tự phân cách `sep`
- Mặc định (`sep=None`) — tách theo **khoảng trắng** và bỏ qua khoảng trắng thừa
- `s.split(sep, maxsplit)` — giới hạn số lần tách

> `split()` trả về **list của string**, không tự chuyển kiểu.

````python
# Tách theo khoảng trắng (mặc định)
cau = "Python rất thú vị"
tu = cau.split()
print(tu)  # ['Python', 'rất', 'thú', 'vị']
````

````python
# Tách theo ký tự phân cách tùy chọn
csv = "An,Bình,Chi,Hieu"
ten = csv.split(",")
print(ten)  # ['An', 'Bình', 'Chi', 'Hieu']

path = "usr/local/bin/python"
parts = path.split("/")
print(parts)  # ['usr', 'local', 'bin', 'python']
````

````python
# maxsplit — giới hạn số lần tách
s = "a:b:c:d:e"
print(s.split(":", 2))  # ['a', 'b', 'c:d:e'] — tách tối đa 2 lần

# Kết hợp với map() để chuyển sang số nguyên
dong = "1 2 3 4 5"
so = list(map(int, dong.split()))
print(so)   # [1, 2, 3, 4, 5]
print(type(so[0]))  # <class 'int'>
````

## List → Chuỗi — `str.join()`

- `sep.join(iterable)` — nối các phần tử của `iterable` thành 1 chuỗi, ngăn cách bằng `sep`
- **Tất cả phần tử phải là string** — nếu không sẽ báo `TypeError`
- Dùng `map(str, ds)` để chuyển đổi khi list chứa số

> `join()` ngược với `split()` — hai hàm thường đi cặp với nhau.

````python
# Nối với dấu phân cách khác nhau
ten = ["An", "Bình", "Chi"]

print(", ".join(ten))   # An, Bình, Chi
print(" - ".join(ten))  # An - Bình - Chi
print("".join(ten))     # AnBìnhChi (không có dấu cách)
````

````python
# List chứa số → phải map(str, ...) trước
so = [1, 2, 3, 4, 5]
# print(", ".join(so))  # TypeError: sequence item 0: expected str instance, int found

chuoi = ", ".join(map(str, so))
print(chuoi)  # 1, 2, 3, 4, 5
````

````python
# split() và join() đi cặp — round-trip
cau_goc = "Python rất thú vị"
tu = cau_goc.split()          # tách thành list
cau_moi = " ".join(tu)        # nối lại
print(cau_moi == cau_goc)     # True

# Ứng dụng: đảo thứ tự từ trong câu
dao_nguoc = " ".join(reversed(cau_goc.split()))
print(dao_nguoc)  # vị thú rất Python
````

## Thay Thế Chuỗi — `str.replace()`

- `s.replace(old, new)` — thay **tất cả** lần xuất hiện của `old` bằng `new`
- `s.replace(old, new, count)` — giới hạn số lần thay thế
- **Không thay đổi chuỗi gốc** — trả về chuỗi mới (string là immutable)

> Dùng `replace()` khi biết chính xác ký tự/chuỗi cần thay. Dùng `re.sub()` khi cần pattern phức tạp hơn.

````python
# Thay tất cả lần xuất hiện
s = "Python rất hay, Python dễ học"
moi = s.replace("Python", "Java")
print(moi)   # Java rất hay, Java dễ học
print(s)     # Python rất hay, Python dễ học — chuỗi gốc không đổi
````

````python
# count — giới hạn số lần thay thế
s = "aaa bbb aaa ccc aaa"
print(s.replace("aaa", "XXX"))       # XXX bbb XXX ccc XXX — thay hết
print(s.replace("aaa", "XXX", 2))    # XXX bbb XXX ccc aaa — chỉ thay 2 lần đầu
````

````python
# Ứng dụng thực tế
# 1. Xóa ký tự (thay bằng chuỗi rỗng)
so_dien_thoai = "0912-345-678"
chuan = so_dien_thoai.replace("-", "")
print(chuan)  # 0912345678

# 2. Chuỗi hóa số thập phân theo chuẩn địa phương
gia = "1.500.000 đ"
print(gia.replace(".", ","))  # 1,500,000 đ

# 3. Kết hợp nhiều replace liên tiếp (chaining)
van_ban = "  xin   chào   bạn  "
sach = van_ban.strip().replace("   ", " ").replace("  ", " ")
print(repr(sach))  # 'xin chào bạn'
````

## Tìm Kiếm Trong Chuỗi — `find()` và `index()`

| | `find()` | `index()` |
|---|---|---|
| Tìm thấy | Trả về **index đầu tiên** | Trả về **index đầu tiên** |
| Không tìm thấy | Trả về **`-1`** | Báo **`ValueError`** |
| Dùng khi | Không chắc có tồn tại | Chắc chắn tồn tại |

Cú pháp đầy đủ: `s.find(sub, start, end)` — tìm trong đoạn `s[start:end]`

> Cả hai đều tìm **lần xuất hiện đầu tiên**. Dùng `rfind()` / `rindex()` để tìm từ cuối.

````python
# find() — trả về -1 nếu không tìm thấy
s = "Python rất hay, Python dễ học"

print(s.find("Python"))   # 0  — index đầu tiên
print(s.find("Java"))     # -1 — không tìm thấy, không báo lỗi
print(s.find("hay"))      # 11
````

````python
# index() — báo ValueError nếu không tìm thấy
s = "Python rất hay, Python dễ học"

print(s.index("Python"))  # 0
print(s.index("hay"))     # 11
# print(s.index("Java"))  # ValueError: substring not found
````

````python
# Tìm trong đoạn [start:end] và tìm từ cuối
s = "Python rất hay, Python dễ học"

print(s.find("Python", 1))      # 16 — bỏ qua lần đầu, tìm từ index 1 trở đi
print(s.rfind("Python"))        # 16 — tìm từ cuối về đầu
print(s.rindex("Python"))       # 16
````

````python
# Khi nào dùng find() vs index()
email = "user@example.com"

# find() — kiểm tra trước khi xử lý
if email.find("@") != -1:
    vi_tri = email.find("@")
    domain = email[vi_tri + 1:]
    print(f"Domain: {domain}")  # Domain: example.com

# index() — khi chắc chắn dữ liệu hợp lệ
try:
    vi_tri = email.index("@")
    print(f"@ ở vị trí: {vi_tri}")  # @ ở vị trí: 4
except ValueError:
    print("Email không hợp lệ")
````
