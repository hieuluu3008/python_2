# Tuple — Bộ Dữ Liệu Bất Biến

**Đặc điểm:** Immutable (bất biến), ordered, indexed, cho phép phần tử trùng lặp.

## Tạo Tuple

Dùng `()` hoặc hàm `tuple()`. Tuple 1 phần tử **bắt buộc có dấu phẩy** sau giá trị.

```python
my_tuple = (1, "hello", 3.14)
single_element = (99,)          # dấu phẩy bắt buộc / nếu không có dấu "," sẽ thành int
tuple_from_list = tuple([1, 2, 3])
print(my_tuple)
print(single_element)
print(tuple_from_list)
```

## Truy Cập Phần Tử — Index & Slicing

- `t[i]` — truy cập theo index, bắt đầu từ `0`
- `t[-1]` — index âm đếm từ cuối
- `t[start:end]` — slicing, lấy từ `start` đến `end-1`

```python
coordinates = (10, 20, 30)
print(coordinates[0])    # 10
print(coordinates[-1])   # 30
print(coordinates[0:2])  # (10, 20)
```

## Tuple Unpacking

Gán từng phần tử của tuple vào các biến riêng biệt. Số biến phải khớp với số phần tử.

```python
point = (10, 25, 35)
x, y, z = point
print(f"x = {x}, y = {y}")  # x = 10, y = 25
```

```python
# Unpacking với * (star expression)
first, *rest = (1, 2, 3, 4, 5)
print(first)  # 1
print(rest)   # [2, 3, 4, 5] LIST
```

## Methods & Hàm Tiện Ích

Tuple chỉ có 2 method (vì immutable):
- `count(value)` — đếm số lần xuất hiện của `value`
- `index(value)` — trả về index đầu tiên của `value`

Ngoài ra dùng `len()` để lấy độ dài.

```python
t = (1, 2, 2, 3, 4, 2)
print(len(t))         # 6
print(t.count(2))     # 3
print(t.index(3))     # 3
```

## Toán Tử Với Tuple

- `+` — nối 2 tuple tạo tuple mới
- `*` — lặp tuple
- `in` — kiểm tra phần tử có trong tuple không

> Các toán tử này **không sửa tuple gốc**, chúng tạo ra tuple mới.

```python
t1 = (1, 3)
t2 = (2, 4)
t_noi = t1 + t2
t_lap = t1 * 2
print(t_noi)         # (1, 2, 3, 4)
print(t_lap)         # (1, 2, 1, 2)
print(5 in t_noi)    # False
print(3 in t_noi)    # True
```

## Tuple vs List — Khi Nào Dùng Tuple?

| Tiêu chí | Tuple | List |
|----------|-------|------|
| Thay đổi được | Không | Có |
| Dùng làm key dict | Có | Không |
| Tốc độ | Nhanh hơn | Chậm hơn |
| Ý nghĩa | Dữ liệu cố định | Dữ liệu động |

**Dùng tuple khi:**
- Dữ liệu không nên thay đổi (tọa độ, màu RGB, cấu hình)
- Cần làm key cho dictionary
- Hàm trả về nhiều giá trị

```python
# Tuple làm key dictionary
locations = {
    (10.762622, 106.660172): "TP.HCM",
    (21.027764, 105.834160): "Hà Nội",
}
print(locations[(10.762622, 106.660172)])  # TP.HCM
```

```python
# Hàm trả về nhiều giá trị dưới dạng tuple
def min_max(ds):
    return min(ds), max(ds)

lo, hi = min_max([3, 1, 4, 1, 5, 9, 2])
print(f"Min: {lo}, Max: {hi}")  # Min: 1, Max: 9
```

### Luyện Tập

- **Bài 1:** Cho tuple `diem = (7, 8, 5, 9, 6, 4, 8, 10, 3)`. Đếm số lần xuất hiện của điểm `8`, tìm vị trí xuất hiện đầu tiên của điểm `5`, và tính tổng tất cả các điểm.
- **Bài 2:** Viết hàm `thong_ke(ds)` nhận vào một list số, trả về tuple `(min, max, trung_binh)`. Unpack kết quả ra 3 biến riêng và in ra.
- **Bài 3:** Có dict `{"An": 8.5, "Bình": 7.0, "Chi": 9.0}`. Dùng `.items()` để lấy các cặp key-value dưới dạng tuple, in ra từng cặp `(tên, điểm)`.

```python
# Bài 1
diem = (7,8,5,9,6,4,8,10,3)
print(diem.count(8))
print(diem.index(5))
print(sum(diem))
```

```python
# Bài 2
def thong_ke(ds):
    return min(ds), max(ds), sum(ds) / len(ds)
min, max, trung_binh = thong_ke(diem)
print(f"Min: {min}, Max: {max}, Trung binh: {trung_binh}")
```

```python
# Bài 3
dict = {"An": 8.5, "Bình": 7.0, "Chi": 9.0}
for a in dict.items():
    print(a)
```
