Bạn đang xây dựng backend cho một trang thương mại điện tử.

**Dữ liệu ban đầu:**
```python
gio_hang = [
    {"ten": "Áo thun",    "gia": 150000, "so_luong": 2,  "danh_muc": "quan_ao"},
    {"ten": "Quần jean",  "gia": 450000, "so_luong": 1,  "danh_muc": "quan_ao"},
    {"ten": "Tai nghe",   "gia": 800000, "so_luong": 1,  "danh_muc": "dien_tu"},
    {"ten": "Sạc dự phòng","gia": 350000, "so_luong": 2, "danh_muc": "dien_tu"},
    {"ten": "Bình nước",  "gia": 120000, "so_luong": 3,  "danh_muc": "gia_dung"},
    {"ten": "Đèn LED",    "gia": 75000,  "so_luong": 4,  "danh_muc": "gia_dung"},
]
```

**Yêu cầu:**

**a)** Viết hàm `them_san_pham(gio_hang, ten, gia, so_luong=1, danh_muc="khac")` thêm sản phẩm mới vào giỏ. Nếu sản phẩm (cùng `ten`) đã tồn tại thì **cộng thêm** số lượng thay vì thêm mới.

**b)** Viết hàm `tinh_tong_tien(gio_hang, *ma_giam_gia, **tuy_chon)` tính tổng tiền:
- `ma_giam_gia`: mỗi mã là một số nguyên % giảm (ví dụ `10` = giảm 10%). Áp dụng **tất cả** mã giảm giá cộng lại
- `**tuy_chon`: hỗ trợ `phi_ship=0` (phí vận chuyển) và `danh_muc=None` (nếu có, chỉ tính sản phẩm thuộc danh mục đó)

**c)** Viết hàm `xuat_bao_cao(gio_hang)` trả về dict gồm:
- `"tong_san_pham"`: tổng số loại sản phẩm
- `"tong_tien_truoc_giam"`: tổng tiền chưa giảm
- `"dat_nhat"`: tên sản phẩm có đơn giá cao nhất
- `"re_nhat"`: tên sản phẩm có đơn giá thấp nhất
- `"theo_danh_muc"`: dict `{danh_muc: tong_tien}` cho từng danh mục

````python
gio_hang = [
    {"ten": "Áo thun",     "gia": 150000, "so_luong": 2, "danh_muc": "quan_ao"},
    {"ten": "Quần jean",   "gia": 450000, "so_luong": 1, "danh_muc": "quan_ao"},
    {"ten": "Tai nghe",    "gia": 800000, "so_luong": 1, "danh_muc": "dien_tu"},
    {"ten": "Đèn LED",     "gia": 75000,  "so_luong": 4, "danh_muc": "gia_dung"},
]
````

````python
# a) Viết hàm them_san_pham
def them_san_pham(gio_hang, ten, gia, so_luong=1, danh_muc="khac"):
    if ten in [sp["ten"] for sp in gio_hang]:
        for sp in gio_hang:
            if sp["ten"] == ten:
                sp["so_luong"] += so_luong
                break 
    else:
        gio_hang.append(dict(ten=ten,gia=gia,so_luong=so_luong,danh_muc=danh_muc))
them_san_pham(gio_hang=gio_hang,ten="Áo thun",gia=150000, so_luong=3, danh_muc="quna_ao")
them_san_pham(gio_hang=gio_hang,ten="Áo sơ mi",gia=200000, so_luong=3, danh_muc="quna_ao")
````

````python
# b) Viết hàm tinh_tong_tien
def tinh_tong_tien(gio_hang, *ma_giam_gia, **tuy_chon):
    phi_ship = tuy_chon.get("phi_ship",0)
    danh_muc = tuy_chon.get("danh_muc",None)
    tong_tien = []
    for sp in gio_hang:
        gia_tien = sp["gia"] * sp["so_luong"]
        tong_tien.append(gia_tien)
    bill = sum(tong_tien) * (1- sum(ma_giam_gia) / 100)
    danh_muc_chiu_phi = ["Đèn LED","Tai nghe"]
    if danh_muc in danh_muc_chiu_phi:
        bill += phi_ship
    print(bill)
tinh_tong_tien(gio_hang,10,20,phi_ship = 10000, danh_muc="Tai nghe")
````

````python
# c) Viết hàm xuat_bao_cao
def xuat_bao_cao(don_hang):
    tong_san_pham = 0
    tong_tien_truoc_giam = 0 
    theo_danh_muc = {} 
    for sp in don_hang:
        tong_san_pham += 1
        tien_san_pham = sp["gia"] * sp["so_luong"]
        tong_tien_truoc_giam += tien_san_pham
        dm = sp["danh_muc"]
        theo_danh_muc[dm] = theo_danh_muc.get(dm, 0) + tien_san_pham
    dat_nhat = max(gio_hang, key=lambda x: x["gia"])["ten"]
    re_nhat =  min(gio_hang, key=lambda x: x["gia"])["ten"]
    tao_dict = dict(tong_san_pham=tong_san_pham, tong_tien_truoc_giam=tong_tien_truoc_giam, dat_nhat=dat_nhat, re_nhat=re_nhat, theo_danh_muc= theo_danh_muc)
    # tao_dict = dict(danh_muc)
    print(tao_dict)

xuat_bao_cao(don_hang=gio_hang)
    

    
````

## Bài 2 — Quản Lý Bảng Điểm Lớp Học

**Dữ liệu ban đầu:**
```python
lop_hoc = {
    "SV001": {"ten": "Nguyễn An",   "diem": {"Toan": 8.5, "Ly": 7.0, "Hoa": 9.0}},
    "SV002": {"ten": "Trần Bình",   "diem": {"Toan": 5.0, "Ly": 6.5, "Hoa": 4.5}},
    "SV003": {"ten": "Lê Chi",      "diem": {"Toan": 9.5, "Ly": 8.0, "Hoa": 9.5}},
    "SV004": {"ten": "Phạm Dũng",   "diem": {"Toan": 6.0, "Ly": 5.5, "Hoa": 7.0}},
    "SV005": {"ten": "Hoàng Em",    "diem": {"Toan": 3.5, "Ly": 4.0, "Hoa": 5.0}},
}
```

**Yêu cầu:**

**a)** Viết hàm `cap_nhat_diem(lop_hoc, ma_sv, **diem_moi)` cập nhật điểm cho sinh viên. Nếu `ma_sv` không tồn tại, in ra cảnh báo và không làm gì. Hỗ trợ cập nhật một hoặc nhiều môn cùng lúc.

**b)** Viết hàm `xep_loai(trung_binh)` trả về xếp loại:
- `>= 8.5` → `"Xuất sắc"` | `>= 7.0` → `"Giỏi"` | `>= 5.5` → `"Khá"` | `>= 4.0` → `"Trung bình"` | còn lại → `"Yếu"`

**c)** Viết hàm `tong_ket_lop(lop_hoc, *mon_tinh, he_so=None)` tổng kết lớp:
- `mon_tinh`: chỉ tính điểm TB từ các môn được liệt kê (nếu rỗng → tính tất cả môn)
- `he_so`: dict `{mon: trong_so}` để tính điểm TB có trọng số (nếu `None` → trọng số đều nhau)
- Trả về list các dict `{"ma_sv", "ten", "trung_binh", "xep_loai"}` sắp xếp theo điểm TB **giảm dần**

**d)** *(Nâng cao)* Viết hàm `thong_ke_mon(lop_hoc, mon)` trả về dict:
- `"cao_nhat"`, `"thap_nhat"`, `"trung_binh"`, `"do_lech_chuan"` (population std)
- Nếu `mon` không tồn tại trong dữ liệu, raise `ValueError` với message rõ ràng

````python
lop_hoc = {
    "SV001": {"ten": "Nguyễn An",  "diem": {"Toan": 8.5, "Ly": 7.0, "Hoa": 9.0}},
    "SV002": {"ten": "Trần Bình",  "diem": {"Toan": 5.0, "Ly": 6.5, "Hoa": 4.5}},
    "SV003": {"ten": "Lê Chi",     "diem": {"Toan": 9.5, "Ly": 8.0, "Hoa": 9.5}},
    "SV004": {"ten": "Phạm Dũng",  "diem": {"Toan": 6.0, "Ly": 5.5, "Hoa": 7.0}},
    "SV005": {"ten": "Hoàng Em",   "diem": {"Toan": 3.5, "Ly": 4.0, "Hoa": 5.0}},
}
````

````python
# a) cap_nhat_diem
def cap_nhat_diem(lop_hoc, ma_sv, **diem_moi):
    if ma_sv not in lop_hoc:
        print(f"Cảnh báo: Mã sinh viên '{ma_sv}' không tồn tại!")
        return
    lop_hoc[ma_sv]["diem"].update(diem_moi)
    print(lop_hoc)
cap_nhat_diem(lop_hoc, "SV001",Toan = 9.0, Ly = 10.0)
````

````python
# b) xep_loai
def xep_loai(trung_binh):
    if trung_binh >= 8.5:
        danh_hieu = "Xuất sắc"
    elif trung_binh >= 7.0:
        danh_hieu = "Giỏi"
    elif trung_binh >= 5.5:
        danh_hieu = "Khá"
    elif trung_binh >= 4.0:
        danh_hieu = "Trung bình"
    else:
        danh_hieu = "Yếu"
    return danh_hieu
````

````python
# c) tong_ket_lop
def tong_ket_lop(lop_hoc, *mon_tinh, he_so=None):
    ket_qua = []
    for ma_sv, tt in lop_hoc.items():
        if mon_tinh:
            mon_hoc = [m for m in tt["diem"].keys() if m in mon_tinh]
        else:
            mon_hoc = [m for m in tt["diem"].keys()]
            
        if he_so:
            tong_diem = sum(tt["diem"][m]*he_so.get(m,1) for m in mon_hoc)
            tong_trong_so = sum(he_so.get(m,1) for m in mon_hoc)
            trung_binh = tong_diem / tong_trong_so
        else:
            tong_diem = sum(tt["diem"][m] for m in mon_hoc)
            trung_binh = tong_diem / len(mon_hoc)
        xep_hang = xep_loai(trung_binh)
        ket_qua.append({"ma_sv":ma_sv, "ten":tt["ten"], "trung_binh": trung_binh, "xep_loai": xep_hang})
    return ket_qua

tong_ket_lop(lop_hoc,"Toan","Hoa")
````

````python
# d) thong_ke_mon
import numpy as np
def thong_ke_mon(lop_hoc, mon):
    diem = []
    for ma_sv, tt in lop_hoc.items():
        diem.append(tt["diem"][mon])
    cao_nhat = max(diem)
    thap_nhat = min(diem)
    trung_binh = sum(diem) / len(diem)
    do_lech_chuan = float(np.std(diem))
    ket_qua = {"cao_nhat": cao_nhat, "thap_nhat": thap_nhat, "trung_binh": trung_binh, "do_lech_chuan": do_lech_chuan}
    return ket_qua
thong_ke_mon(lop_hoc,"Toan")
````

## Bài 3 — Hệ Thống Cấu Hình Ứng Dụng

Mô phỏng cơ chế load config theo lớp (layered config): `default` → `env` → `user` (sau ghi đè trước).

**Yêu cầu:**

**a)** Viết hàm `merge_config(*configs)` nhận nhiều dict config, merge theo thứ tự (dict sau **ghi đè** dict trước). Với các value là dict, thực hiện **deep merge** (không ghi đè toàn bộ, mà merge từng key con).

Ví dụ:
```python
default  = {"db": {"host": "localhost", "port": 5432}, "debug": False}
env      = {"db": {"port": 5433},                      "debug": True}
# Kết quả mong đợi:
# {"db": {"host": "localhost", "port": 5433}, "debug": True}
```

**b)** Viết hàm `validate_config(config, **bat_buoc)` kiểm tra config:
- `**bat_buoc`: mỗi key là tên field, value là kiểu dữ liệu mong đợi (ví dụ `host=str, port=int`)
- Hỗ trợ kiểm tra field lồng nhau dùng dấu chấm: `"db.host"=str`
- Trả về list các lỗi (string), list rỗng nếu hợp lệ

**c)** Viết hàm `in_config(config, indent=0, **bo_loc)` in config dạng cây:
- `indent`: số space thụt đầu dòng (tự tăng khi gặp dict lồng)
- `**bo_loc`: nếu có `an_key=["password", "secret"]`, thay giá trị của các key đó bằng `"***"`

````python
# a) merge_config — deep merge
default  = {"db": {"host": "localhost", "port": 5432}, "debug": False}
env      = {"db": {"port": 5433},                      "debug": True}


def merge_config(*configs):
    config = {}
    for cfg in configs:                          # lần lượt từng config truyền vào
        for k, v in cfg.items():
            if k in config and isinstance(config[k], dict) and isinstance(v, dict):
                config[k] = {**config[k], **v}  # deep merge nếu cả 2 là dict
            else:
                config[k] = v                   # ghi đè bình thường
    return config

merge_config(default,env)
````

    {"id": "DH003", "khach": "An",   "san_pham": "Giày",  "so_luong": 1, "don_gia": 500000, "thang": 2},
    {"id": "DH004", "khach": "Chi",  "san_pham": "Áo",    "so_luong": 3, "don_gia": 150000, "thang": 2},
    {"id": "DH005", "khach": "Bình", "san_pham": "Giày",  "so_luong": 2, "don_gia": 500000, "thang": 3},

**Yêu cầu:**

**a)** Viết hàm `doanh_thu_theo(don_hang, nhom_by, *thang_loc)` phân tích doanh thu:
- `nhom_by`: `"khach"` hoặc `"san_pham"` hoặc `"thang"` — nhóm kết quả theo trường này
- `thang_loc`: nếu có, chỉ tính các tháng được liệt kê
- Trả về dict `{nhom: tong_doanh_thu}` sắp xếp theo doanh thu **giảm dần**

**b)** Viết hàm `top_khach_hang(don_hang, top_n=3, **tieu_chi)` trả về top N khách hàng:
- `**tieu_chi` hỗ trợ `theo="doanh_thu"` (mặc định) hoặc `theo="so_don"` (số lần mua)
- Nếu `theo` là giá trị không hợp lệ, raise `ValueError`
- Trả về list `[(ten_khach, gia_tri)]` đã sắp xếp

**c)** *(Nâng cao)* Viết hàm `du_bao_thang_sau(don_hang, phuong_phap="trung_binh")` dự báo doanh thu tháng tiếp theo:
- `"trung_binh"`: trung bình cộng doanh thu các tháng
- `"xu_huong"`: dùng doanh thu 2 tháng gần nhất để nội suy tuyến tính
- Nếu `phuong_phap` không hợp lệ, raise `ValueError`

````python
don_hang = [
    {"id": "DH001", "khach": "An",   "san_pham": "Áo",   "so_luong": 2, "don_gia": 150000, "thang": 1},
    {"id": "DH002", "khach": "Bình", "san_pham": "Quần", "so_luong": 1, "don_gia": 300000, "thang": 1},
    {"id": "DH003", "khach": "An",   "san_pham": "Giày", "so_luong": 1, "don_gia": 500000, "thang": 2},
    {"id": "DH004", "khach": "Chi",  "san_pham": "Áo",   "so_luong": 3, "don_gia": 150000, "thang": 2},
    {"id": "DH005", "khach": "Bình", "san_pham": "Giày", "so_luong": 2, "don_gia": 500000, "thang": 3},
    {"id": "DH006", "khach": "An",   "san_pham": "Quần", "so_luong": 2, "don_gia": 300000, "thang": 3},
    {"id": "DH007", "khach": "Chi",  "san_pham": "Túi",  "so_luong": 4, "don_gia": 200000, "thang": 3},
    {"id": "DH008", "khach": "Dũng", "san_pham": "Áo",   "so_luong": 5, "don_gia": 150000, "thang": 4},
]
````

````python
# a) doanh_thu_theo
def doanh_thu_theo(don_hang, nhom_by, *thang_loc):
    doanh_thu_theo_nhom = {}
    if thang_loc:
        filter_thang = [i["thang"] for i in don_hang if i["thang"] in thang_loc]
    else:
        filter_thang = [i["thang"] for i in don_hang]
    for i in don_hang:
        if i["thang"] in filter_thang:
            doanh_thu = i["don_gia"] * i["so_luong"]
            if i[nhom_by] in doanh_thu_theo_nhom:
                doanh_thu_theo_nhom[i[nhom_by]] += doanh_thu
            else:
                doanh_thu_theo_nhom[i[nhom_by]] = doanh_thu
    return dict(sorted(doanh_thu_theo_nhom.items(), key=lambda x: x[1], reverse=True))
doanh_thu_theo(don_hang, "san_pham",1)
````

````python
# b) top_khach_hang
def top_khach_hang(don_hang, top_n=3, **tieu_chi):
    theo = tieu_chi.get("theo","doanh_thu")
    if theo not in ["doanh_thu", "so_don"]:
        raise ValueError(f"Tiêu chí {theo} không hợp lệ.")
    top_khach = {}
    for dh in don_hang:
        khach = dh["khach"]
        if theo == "doanh_thu":
            gia_tri = dh["don_gia"] * dh["so_luong"]
        else:
            gia_tri = 1 
        top_khach[khach] = top_khach.get(khach,0) + gia_tri
    return sorted(top_khach.items(), key= lambda x: x[1], reverse= True)[:top_n]
top_khach_hang(don_hang,top_n=2,theo="doanh_thu")
````
