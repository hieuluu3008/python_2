# Dict — Từ Điển

**Đặc điểm:** Mutable, ordered (Python 3.7+), truy cập qua key, key phải unique và immutable.

## Tạo Dict

- `{}` với cặp `key: value`
- `dict()` với keyword arguments
- `dict()` từ list of tuples

```python
student = {
    "name": "Nguyễn Văn A",
    "id": "SV001",
    "age": 20,
    "gpa": 8.5
}
print(student)
```

```python
info = dict(product="Item X", price=150.0, stock=50)
print(info)
```

```python

# pairs = [("color", "blue"), ("size", "L")] #Mỗi tuple trong list là một cặp key-value
pairs = [["color", "blue"], ["size", "L"]]   #Mỗi list con trong list là một cặp key-value
d = dict(pairs)
print(d)
```

## Truy Cập Giá Trị — `[]` / `get()`

- `dict[key]` — truy cập trực tiếp, báo `KeyError` nếu key không tồn tại
- `dict.get(key)` — trả về `None` nếu không tìm thấy
- `dict.get(key, default)` — trả về `default` nếu không tìm thấy

> Dùng `get()` khi không chắc key có tồn tại để tránh lỗi.

```python
student = {"name": "Nguyễn Văn A", "age": 21}

print(student["name"])                             # Nguyễn Văn A
print(student.get("age"))                          # 21
print(student.get("department", "No data"))  # No data
```

## Thêm & Cập Nhật — `[]` / `update()`

- `dict[key] = value` — thêm mới nếu key chưa có, cập nhật nếu key đã có
- `update({...})` — cập nhật nhiều cặp key-value cùng lúc

```python
student = {"name": "Nguyễn Văn A"}

student["age"] = 21
student["id"] = "SV001"
print(student)  # {'name': 'Nguyễn Văn A', 'age': 21, 'id': 'SV001'}
```

```python
student.update({"gpa": 8.0, "class": "K60", "age": 22})
print(student)
```

## Xóa Phần Tử — `pop` / `popitem` / `del` / `clear`

- `pop(key)` — xóa và trả về giá trị; báo `KeyError` nếu không có
- `pop(key, default)` — trả về `default` thay vì lỗi nếu key không tồn tại
- `popitem()` — xóa và trả về cặp key-value được chèn **cuối cùng** (Python 3.7+)
- `del dict[key]` — xóa theo key
- `clear()` — xóa toàn bộ

```python
config = {"host": "localhost", "port": 8080, "debug": True}

port = config.pop("port")
print(port)    # 8080
print(config)  # {'host': 'localhost', 'debug': True}
```

```python
config = {"host": "localhost", "port": 8080, "debug": True}

safe = config.pop("db", None)
print(safe)  # None — không báo lỗi
```

```python
config = {"host": "localhost", "port": 8080, "debug": True}

last = config.popitem()
print(last)    # ('debug', True)
print(config)  # {'host': 'localhost', 'port': 8080}
```

```python
config = {"host": "localhost", "port": 8080, "debug": True}

del config["debug"]
print(config)  # {'host': 'localhost', 'port': 8080}

config.clear()
print(config)  # {}
```

## Duyệt Dict — `keys` / `values` / `items`

- `for key in dict` — duyệt qua các key
- `dict.values()` — duyệt qua các value
- `dict.items()` — duyệt qua từng cặp `(key, value)`

```python
student = {"name": "Nguyễn Văn A", "age": 21, "id": "SV001"}

# for key in student:
#     print(key)

for key in student.keys():
    print(key)
```

```python
for value in student.values():
    print(value)
```

```python
for key, value in student.items():
    print(f"{key}: {value}")
```

## Tiện Ích — `len` / `in` / `copy`

- `len(dict)` — số lượng cặp key-value
- `key in dict` — kiểm tra key có tồn tại không
- `value in dict.values()` — kiểm tra value
- `dict.copy()` — tạo bản sao nông (shallow copy), thay đổi bản sao không ảnh hưởng bản gốc

```python
scores = {"Alice": 90, "Bob": 85}

print(len(scores))              # 2
print("Alice" in scores)        # True
print("Charlie" in scores)      # False
print(85 in scores.values())    # True
```

```python
scores_copy = scores.copy()
scores_copy["Alice"] = 100

print(scores_copy)  # {'Alice': 100, 'Bob': 85}
print(scores)       # {'Alice': 90, 'Bob': 85} — bản gốc không đổi
```
