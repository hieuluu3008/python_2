# Selection Sort (Sắp xếp Chọn)


## 1. Định nghĩa

**Selection Sort** (Sắp xếp Chọn) là một thuật toán sắp xếp đơn giản, chia danh sách thành **hai phần**:

- **Phần đã sắp xếp** — nằm bên trái, phát triển dần từng bước.
- **Phần chưa sắp xếp** — nằm bên phải, thu hẹp dần.

Thuật toán hoạt động bằng cách **tìm phần tử nhỏ nhất** trong phần chưa sắp xếp và đưa nó về đúng vị trí.

## 2. Cách hoạt động

Quy trình lặp lại 4 bước:

1. Tìm phần tử **nhỏ nhất** trong phần chưa sắp xếp.
2. **Hoán đổi** nó với phần tử đầu tiên của phần chưa sắp xếp.
3. **Mở rộng** phần đã sắp xếp sang phải thêm 1 vị trí.
4. **Thu hẹp** phần chưa sắp xếp tương ứng.

### Minh họa với `[64, 25, 12, 22, 11]`

```
Mảng ban đầu:  [64, 25, 12, 22, 11]
                ^--- phần chưa sắp xếp ---^

Lượt 1: min = 11 (index 4) → swap(0, 4) → [11 | 25, 12, 22, 64]
Lượt 2: min = 12 (index 2) → swap(1, 2) → [11, 12 | 25, 22, 64]
Lượt 3: min = 22 (index 3) → swap(2, 3) → [11, 12, 22 | 25, 64]
Lượt 4: min = 25 (index 3) → swap(3, 3) → [11, 12, 22, 25 | 64]
Kết quả:       [11, 12, 22, 25, 64] ✓
```

> **Lưu ý quan trọng:** Selection Sort luôn thực hiện đúng **n-1 lượt**, bất kể dữ liệu đầu vào là gì — không có cơ chế early termination như Bubble Sort.

## 3. Các bước thuật toán

```
1. Lấy độ dài mảng n.
2. Vòng lặp ngoài i từ 0 đến n-2:
   a. Gán min_index = i (giả sử phần tử tại i là nhỏ nhất).
   b. Vòng lặp trong j từ i+1 đến n-1:
      - Nếu arr[j] < arr[min_index] → cập nhật min_index = j.
   c. Hoán đổi arr[i] và arr[min_index].
3. Mảng đã được sắp xếp.
```

## 4. Cài đặt Python

````python
def selection_sort(arr: list) -> list:
    """Sắp xếp danh sách theo thứ tự tăng dần bằng Selection Sort.

    Args:
        arr: Danh sách các phần tử có thể so sánh được.

    Returns:
        Danh sách đã được sắp xếp (in-place).
    """
    n = len(arr)

    for i in range(n - 1):           # n-1 lượt là đủ
        min_index = i                # Giả sử vị trí i là nhỏ nhất

        for j in range(i + 1, n):   # Tìm phần tử nhỏ nhất trong phần chưa sắp xếp
            if arr[j] < arr[min_index]:
                min_index = j

        # Chỉ hoán đổi nếu tìm được vị trí nhỏ hơn (tránh swap thừa)
        if min_index != i:
            arr[i], arr[min_index] = arr[min_index], arr[i]

    return arr
````

````python
# --- Chạy thử với ví dụ từ bài học ---
data = [64, 25, 12, 22, 11]
print("Trước khi sắp xếp:", data)

sorted_data = selection_sort(data.copy())
print("Sau khi sắp xếp: ", sorted_data)
````

````python
# --- Minh họa từng bước ---
def selection_sort_verbose(arr: list) -> list:
    """Selection Sort với in từng bước để minh họa."""
    n = len(arr)
    arr = arr.copy()

    for i in range(n - 1):
        min_index = i
        for j in range(i + 1, n):
            if arr[j] < arr[min_index]:
                min_index = j

        if min_index != i:
            arr[i], arr[min_index] = arr[min_index], arr[i]
            print(f"Lượt {i + 1}: min={arr[i]} (swap vị trí {i} ↔ {min_index}) → {arr}")
        else:
            print(f"Lượt {i + 1}: min={arr[i]} (không cần swap)          → {arr}")

    return arr


selection_sort_verbose([64, 25, 12, 22, 11])
````

````python
# --- Test các trường hợp đặc biệt ---
print(selection_sort([]))           # Mảng rỗng   → []
print(selection_sort([1]))          # Một phần tử → [1]
print(selection_sort([3, 1, 2]))    # Ngẫu nhiên  → [1, 2, 3]
print(selection_sort([1, 2, 3]))    # Đã sắp xếp  → [1, 2, 3]
print(selection_sort([3, 2, 1]))    # Ngược chiều → [1, 2, 3]
print(selection_sort([5, 5, 3, 1])) # Trùng nhau  → [1, 3, 5, 5]
````

## 5. Phân tích độ phức tạp

| Trường hợp | Độ phức tạp thời gian | Ghi chú |
|---|---|---|
| **Tốt nhất** (Best) | O(n²) | Luôn duyệt đủ n-1 lượt, không có early termination |
| **Trung bình** (Average) | O(n²) | Hai vòng lặp lồng nhau |
| **Tệ nhất** (Worst) | O(n²) | Không thay đổi theo dữ liệu đầu vào |
| **Không gian** (Space) | O(1) | Sắp xếp tại chỗ (in-place) |

### Số phép toán chính xác

- **So sánh:** n(n-1)/2 (luôn cố định)
- **Hoán đổi:** tối đa **n-1 lần** — đây là điểm mạnh so với Bubble Sort (có thể swap đến n²/2 lần)

## 6. Ưu điểm & Nhược điểm

### Ưu điểm
- **Đơn giản, dễ hiểu** — logic rõ ràng, tốt cho học thuật toán.
- **In-place** — O(1) bộ nhớ phụ.
- **Tối thiểu số lần hoán đổi** — tối đa n-1 swap, hữu ích khi chi phí ghi bộ nhớ cao.
- **Hiệu suất nhất quán** — O(n²) trong mọi trường hợp, dễ dự đoán.

### Nhược điểm
- **Không có early termination** — luôn chạy đủ n-1 lượt dù mảng đã sắp xếp.
- **Not stable** — có thể thay đổi thứ tự tương đối của các phần tử bằng nhau.
- **Không phù hợp dữ liệu lớn** — O(n²) làm chậm đáng kể khi n tăng.

## 7. So sánh Selection Sort vs Bubble Sort

| Tiêu chí | Bubble Sort | Selection Sort |
|---|---|---|
| **Time (Best)** | O(n) ✅ | O(n²) ❌ |
| **Time (Avg/Worst)** | O(n²) | O(n²) |
| **Space** | O(1) | O(1) |
| **Số lần swap** | Nhiều (đến n²/2) | Ít — tối đa n-1 ✅ |
| **Stable?** | ✅ Có | ❌ Không |
| **Early termination** | ✅ Có | ❌ Không |
| **Dùng khi nào** | Dữ liệu gần sắp xếp | Chi phí write cao, cần ít swap |

## 8. So sánh tổng quan với các thuật toán khác

| Thuật toán | Best | Average | Worst | Space | Stable? |
|---|---|---|---|---|---|
| **Selection Sort** | O(n²) | O(n²) | O(n²) | O(1) | ❌ |
| Bubble Sort | O(n) | O(n²) | O(n²) | O(1) | ✅ |
| Insertion Sort | O(n) | O(n²) | O(n²) | O(1) | ✅ |
| Merge Sort | O(n log n) | O(n log n) | O(n log n) | O(n) | ✅ |
| Quick Sort | O(n log n) | O(n log n) | O(n²) | O(log n) | ❌ |
| Tim Sort (Python) | O(n) | O(n log n) | O(n log n) | O(n) | ✅ |

## 9. Khi nào nên dùng?

- **Học thuật toán** — hiểu rõ khái niệm tìm min và hoán đổi.
- **Bộ nhớ flash / EEPROM** — nơi chi phí ghi (write) rất cao; Selection Sort ghi tối đa n-1 lần.
- **Dữ liệu rất nhỏ** (< 20 phần tử) — overhead của thuật toán phức tạp hơn không đáng.
- ❌ **Không dùng** cho dữ liệt lớn — hãy dùng `sorted()` hoặc `.sort()` của Python (Tim Sort).
