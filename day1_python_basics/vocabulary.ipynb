## 1. Code Basics

### comment
**Định nghĩa:** A line (or part of a line) of code written in English (or another language) preceded by a special symbol that lets the programming language know it should ignore that line (or part of a line) of code.

**Giải thích:** Dòng (hoặc một phần dòng) code được đánh dấu bằng ký hiệu đặc biệt để Python bỏ qua khi chạy. Dùng để giải thích code cho người đọc.

**Ký hiệu:** `#` trong Python

````python
# Đây là comment — Python sẽ bỏ qua dòng này
x = 10  # comment cũng có thể đặt cuối dòng code
print(x)
````

### keyword
**Định nghĩa:** A word with a special meaning in Python.

**Giải thích:** Từ có ý nghĩa đặc biệt trong Python, không thể dùng làm tên biến hay hàm.

````python
import keyword
print(keyword.kwlist)  # xem toàn bộ keyword của Python
print("Tổng số keyword:", len(keyword.kwlist))

# Một số keyword phổ biến:
# if, else, elif, for, while, def, class, return, import, True, False, None, and, or, not, in, is
````

## 2. Data & Values

### constant
**Định nghĩa:** A value that never changes.

**Giải thích:** Giá trị không bao giờ thay đổi trong suốt chương trình. Python không có cơ chế constant thực sự — convention dùng **tên viết HOA** để báo hiệu.

````python
PI = 3.14159          # constant — quy ước viết hoa
MAX_RETRY = 3
DB_HOST = "localhost"

print(PI, MAX_RETRY, DB_HOST)
````

### variable
**Định nghĩa:** A name assigned to a value using the assignment operator.

**Giải thích:** Tên được gán cho một giá trị. Biến có thể thay đổi giá trị trong quá trình chạy.

### assignment operator
**Định nghĩa:** The `=` sign.

**Giải thích:** Dấu `=` dùng để gán giá trị cho biến. Khác với `==` là toán tử so sánh bằng.

````python
ten = "An"           # gán chuỗi vào biến 'ten'
tuoi = 22            # gán số nguyên vào biến 'tuoi'
diem = 8.5           # gán số thực

# Gán nhiều biến cùng lúc
x = y = z = 0

# Gán từ nhiều giá trị (unpacking)
a, b, c = 1, 2, 3

print(ten, tuoi, diem)
print(x, y, z)
print(a, b, c)
````

### increment / decrement
**increment:** Increasing the value of a variable — **tăng** giá trị của biến.

**decrement:** Decreasing the value of a variable — **giảm** giá trị của biến.

````python
dem = 0

# Increment
dem = dem + 1   # cách dài
dem += 1        # cách ngắn (augmented assignment)
print("Sau increment:", dem)  # 2

# Decrement
dem -= 1
print("Sau decrement:", dem)  # 1

# Các augmented assignment khác
dem *= 10   # dem = dem * 10
dem //= 3   # dem = dem // 3
print("Sau *=10 và //=3:", dem)
````

## 3. Data Types

### data type
**Định nghĩa:** A category of data.

**Giải thích:** Phân loại dữ liệu. Python tự động nhận diện data type — dùng `type()` để kiểm tra.

### object
**Định nghĩa:** A data value in Python with three properties: an identity, a data type and a value.

**Giải thích:** Mọi thứ trong Python đều là object — có 3 thuộc tính:
- **identity** — địa chỉ bộ nhớ (xem bằng `id()`)
- **data type** — kiểu dữ liệu (xem bằng `type()`)
- **value** — giá trị (xem trực tiếp)

````python
x = 42
print("identity:", id(x))    # địa chỉ bộ nhớ
print("data type:", type(x)) # <class 'int'>
print("value:", x)            # 42
````

### string · integer · float · boolean · nonetype

| Thuật ngữ | Data type | Mô tả | Ví dụ |
|---|---|---|---|
| **string** | `str` | Chuỗi ký tự, bao quanh bởi dấu nháy | `"hello"`, `'An'` |
| **integer** | `int` | Số nguyên | `42`, `-7`, `0` |
| **float** | `float` | Số thực (có phần thập phân) | `3.14`, `-0.5` |
| **boolean** | `bool` | Giá trị logic, chỉ có `True` hoặc `False` | `True`, `False` |
| **nonetype** | `NoneType` | Đại diện cho "không có giá trị", luôn là `None` | `None` |

````python
s = "Xin chào"     # string
i = 100             # integer
f = 9.81            # float
b = True            # boolean
n = None            # nonetype

for val in [s, i, f, b, n]:
    print(f"{repr(val):20} → {type(val).__name__}")
````

## 4. Syntax & Errors

### syntax
**Định nghĩa:** The set of rules, principles, and processes that govern the structure of sentences in a given language, specifically word order.

**Giải thích:** Tập hợp các quy tắc về cách viết code hợp lệ trong Python — thứ tự từ, dấu câu, thụt lề, v.v.

### syntax error
**Định nghĩa:** A fatal programming error caused by violating a programming language's syntax.

**Giải thích:** Lỗi nghiêm trọng do vi phạm quy tắc cú pháp — Python **không thể chạy** chương trình khi có syntax error.

### exception
**Định nghĩa:** A nonfatal programming error.

**Giải thích:** Lỗi xảy ra **khi chạy** (runtime) nhưng không làm sập toàn bộ chương trình — có thể bắt và xử lý bằng `try/except`.

````python
# Exception — bắt được bằng try/except
try:
    ket_qua = 10 / 0          # ZeroDivisionError
except ZeroDivisionError as e:
    print(f"Exception bắt được: {e}")

try:
    so = int("abc")           # ValueError
except ValueError as e:
    print(f"Exception bắt được: {e}")

# SyntaxError — KHÔNG thể bắt bằng try/except vì xảy ra trước khi chạy
# x = (1 + 2   ← thiếu dấu đóng ngoặc → SyntaxError
````

## 5. Operators & Expressions

### operator / operand / expression

| Thuật ngữ | Định nghĩa | Giải thích |
|---|---|---|
| **operator** | Symbols used with operands in an expression | Ký hiệu dùng để thực hiện phép tính |
| **operand** | A value on either side of an operator | Giá trị ở hai bên toán tử |
| **expression** | Code with an operator surrounded by two operands | Tổ hợp toán tử + operand tạo ra một giá trị |

```
   10    +    5
   ↑         ↑
 operand   operand
         ↑
      operator
└──────────────────┘
   expression
```

### arithmetic operator
**Định nghĩa:** A category of operators used in arithmetic expressions.

**Giải thích:** Nhóm toán tử dùng cho các phép tính số học.

| Operator | Ý nghĩa | Ví dụ |
|---|---|---|
| `+` | Cộng | `3 + 2 = 5` |
| `-` | Trừ | `3 - 2 = 1` |
| `*` | Nhân | `3 * 2 = 6` |
| `/` | Chia (float) | `7 / 2 = 3.5` |
| `//` | Chia lấy phần nguyên | `7 // 2 = 3` |
| `%` | Chia lấy phần dư (modulo) | `7 % 2 = 1` |
| `**` | Lũy thừa | `2 ** 3 = 8` |

### order of operations
**Định nghĩa:** A set of rules used in mathematical calculations to evaluate an expression.

**Giải thích:** Thứ tự ưu tiên khi tính toán — Python theo chuẩn toán học (PEMDAS):

1. `()` — Ngoặc đơn
2. `**` — Lũy thừa
3. `*`, `/`, `//`, `%` — Nhân, chia
4. `+`, `-` — Cộng, trừ

````python
print(2 + 3 * 4)      # 14, không phải 20 — nhân trước
print((2 + 3) * 4)    # 20 — ngoặc ưu tiên cao nhất
print(2 ** 3 ** 2)    # 512 — ** kết hợp từ phải: 2**(3**2) = 2**9
print(10 - 4 / 2)     # 8.0 — chia trước
````

### comparison operator
**Định nghĩa:** A category of operators used in expressions that evaluate to either `True` or `False`.

**Giải thích:** Nhóm toán tử so sánh — kết quả luôn là `True` hoặc `False`.

| Operator | Ý nghĩa |
|---|---|
| `==` | Bằng |
| `!=` | Khác |
| `>` | Lớn hơn |
| `<` | Nhỏ hơn |
| `>=` | Lớn hơn hoặc bằng |
| `<=` | Nhỏ hơn hoặc bằng |

### logical operator
**Định nghĩa:** A category of operators that evaluate two expressions and return either `True` or `False`.

**Giải thích:** Nhóm toán tử logic — kết hợp nhiều điều kiện.

| Operator | Ý nghĩa | Kết quả |
|---|---|---|
| `and` | Và | `True` nếu **cả hai** đều `True` |
| `or` | Hoặc | `True` nếu **ít nhất một** là `True` |
| `not` | Phủ định | Đảo ngược `True`/`False` |

````python
tuoi = 20
co_ve = True

# Comparison operators
print(tuoi >= 18)              # True
print(tuoi == 25)              # False

# Logical operators
print(tuoi >= 18 and co_ve)    # True — cả hai True
print(tuoi < 18 or co_ve)      # True — một trong hai True
print(not co_ve)               # False — đảo ngược
````

## 6. Control Flow

### control structure
**Định nghĩa:** A block of code that makes decisions by analyzing the values of variables.

**Giải thích:** Khối code điều hướng luồng thực thi dựa trên giá trị của biến — gồm `if/else`, vòng lặp `for/while`.

### conditional statement
**Định nghĩa:** Code that is able to execute additional code circumstantially.

**Giải thích:** Code có thể thực thi hoặc bỏ qua tùy theo điều kiện — là dạng control structure phổ biến nhất.

### if / elif / else statement

| Thuật ngữ | Định nghĩa | Giải thích |
|---|---|---|
| **if statement** | The first part of an if else statement | Phần đầu — điều kiện chính |
| **else statement** | The second part of an if else statement | Phần cuối — trường hợp còn lại |
| **elif statement** | Statements added to allow additional decisions | Điều kiện bổ sung giữa `if` và `else` |

**if else statement:** A way for programmers to say "if this happens do this, otherwise do that." — Nếu điều này xảy ra thì làm vậy, ngược lại thì làm thế kia.

````python
diem = 7.5

# if / elif / else — control structure
if diem >= 8:           # if statement
    xep_loai = "Giỏi"
elif diem >= 6.5:       # elif statement
    xep_loai = "Khá"
elif diem >= 5:         # có thể thêm nhiều elif
    xep_loai = "Trung bình"
else:                   # else statement
    xep_loai = "Yếu"

print(f"Điểm {diem} → {xep_loai}")
````

## 7. Statements & Clauses

### statement
**Định nghĩa:** A command or a calculation.

**Giải thích:** Một lệnh hoặc phép tính — đơn vị thực thi cơ bản trong Python.

---

### simple statement
**Định nghĩa:** A statement that can be expressed in one line of code.

**Giải thích:** Câu lệnh nằm gọn trong **một dòng** code.

---

### compound statement
**Định nghĩa:** A statement that generally spans multiple lines (but can be written in one line in some circumstances).

**Giải thích:** Câu lệnh thường kéo dài **nhiều dòng** — gồm `if/else`, `for`, `while`, `def`, `class`.

### clause / header / suite

| Thuật ngữ | Định nghĩa | Giải thích |
|---|---|---|
| **clause** | The building blocks of compound statements — made up of a header followed by suite(s) | Khối xây dựng nên compound statement |
| **header** | A line containing a keyword followed by a colon and indented code | Dòng đầu của clause: `keyword ... :` |
| **suite** | A line of code controlled by a header | Các dòng code thụt lề bên trong header |

```python
if diem >= 8:           ← header (keyword + dấu hai chấm)
    print("Giỏi")      ← suite (thụt lề, do header kiểm soát)
    xep_loai = "Giỏi"  ← suite
└──────────────────────┘
         clause
```

````python
# Simple statements — mỗi cái một dòng
x = 5                    # assignment statement
print(x)                 # expression statement
import math              # import statement

# Compound statement — nhiều dòng
for i in range(3):       # header: keyword 'for' + dấu ':'
    print(i)             # suite 1
    x += i               # suite 2

# Compound statement viết 1 dòng (hợp lệ nhưng ít dùng)
if x > 5: print("lớn hơn 5")
````
