### 1. Class · Object · Method

| Thuật ngữ | Là gì | Ví dụ |
|---|---|---|
| **Class** | Bản thiết kế | `class Nguoi` |
| **Object** | Cá thể tạo từ Class | `an = Nguoi()` |
| **Method** | Hàm bên **trong** Class | `def xin_chao(self):` |

> Hàm ngoài class → `def tinh_tong():` &nbsp;&nbsp;|&nbsp;&nbsp; Method trong class → `def xin_chao(self):`

````python
class Nguoi:              # CLASS — bản thiết kế
    def xin_chao(self):   # METHOD — hàm trong class
        print("Xin chào")

an   = Nguoi()            # OBJECT — cá thể tạo từ class
binh = Nguoi()            # OBJECT — cá thể khác
chi  = Nguoi()            # OBJECT — cá thể khác

an.xin_chao()             # object 'an' gọi method
````

### 2. `self` — tại sao cần?

> **Nhiều object dùng chung 1 method → method phải biết đang làm việc với object nào**

```
an.xin_chao()    →  self = an    →  self.ten = "An"   →  "Tôi là An"
binh.xin_chao()  →  self = binh  →  self.ten = "Bình" →  "Tôi là Bình"
```

`self` = object đang gọi method. Python **tự động** truyền vào — lập trình viên không cần truyền.

````python
class Nguoi:
    def __init__(self, ten):
        self.ten = ten

    def xin_chao(self):
        print(f"self là: {self}")    # chính là object đang gọi
        print(f"Tôi là {self.ten}")

an   = Nguoi("An")
binh = Nguoi("Bình")

an.xin_chao()
print()
binh.xin_chao()

# Hai cách gọi này TƯƠNG ĐƯƠNG:
print("\n--- Gọi trực tiếp qua class ---")
Nguoi.xin_chao(an)   # phải truyền object thủ công
````

**Nếu quên `self`:** Python tự động truyền object vào nhưng không có tham số nào để nhận → `TypeError`.

> Tên `self` là **quy ước** theo PEP 8 — có thể đặt tên khác nhưng tuyệt đối không nên.

````python
class Bug:
    def broken():      # ← thiếu self
        print("Hello")

b = Bug()
b.broken()   # ❌ TypeError: broken() takes 0 positional arguments but 1 was given
````

### 3. `__init__` — Constructor

> Chạy **ngay lập tức** khi tạo object → nạp dữ liệu ban đầu vào object

```
Nguoi("An")  →  __init__ tự động chạy  →  self.ten = "An"  →  lưu vào object
```

- `init` = **initialize** = khởi tạo
- `__` (double underscore) = Python nhận diện và tự gọi khi dùng `ClassName(...)`

````python
class Nguoi:
    def __init__(self, ten):
        self.ten = ten   # lưu tên vào object ngay khi tạo

# Python thấy Nguoi("An") → tự hỏi: class có __init__ không?
# Có! → chạy __init__(self=an, ten="An") ngay lập tức
an = Nguoi("An")
print(an.ten)    # "An"
````

`__init__` cũng là method → `self` luôn là tham số **đầu tiên**:

| Khai báo | Hợp lệ? |
|---|---|
| `def __init__(self):` | ✅ |
| `def __init__(self, ten):` | ✅ |
| `def __init__(self, ten, tuoi):` | ✅ |
| `def __init__():` | ❌ Thiếu `self` |

````python
# Giải thích từng dòng chi tiết
class Nguoi:
    def __init__(self, ten):   # self = object vừa tạo, ten = tham số nhận tên
        self.ten = ten         # lưu ten vào object (ten bên phải → self.ten bên trái)

    def xin_chao(self):        # self = object đang gọi method
        print(f"Tôi là {self.ten}")

an   = Nguoi("An")    # __init__(self=an, ten="An")   → an.ten = "An"
binh = Nguoi("Binh")  # __init__(self=binh, ten="Binh") → binh.ten = "Binh"

an.xin_chao()    # self = an   → self.ten = "An"   → "Tôi là An"
binh.xin_chao()  # self = binh → self.ten = "Binh" → "Tôi là Binh"
````

### 4. `self.attribute` vs biến cục bộ

| | Thuộc về | Tồn tại bao lâu |
|---|---|---|
| `toc_do` | Method — biến tạm | Chỉ khi method đang chạy → **mất** sau đó |
| `self.toc_do` | Object — instance attribute | **Mãi** trong object, dùng được ở mọi method |

```
__init__ chạy:     toc_do = 0        ← tạm thời
                   self.toc_do = 0   ← lưu vào object
__init__ kết thúc: toc_do      ❌ mất
                   self.toc_do  ✅ còn → chay() đọc được
```

````python
class XeHoi:
    def __init__(self, hang_xe, mau_xe, toc_do=0):
        self.hang_xe = hang_xe   # lưu vào object → dùng được mọi nơi
        self.mau_xe  = mau_xe
        self.toc_do  = toc_do    # lưu vào object → chay() đọc được

    def chay(self):
        self.toc_do += 50        # lấy self.toc_do của object ra cộng thêm 50
        print(f"Tốc độ hiện tại: {self.toc_do}")

    def thong_tin(self):
        print(f"Hãng xe : {self.hang_xe}")
        print(f"Màu xe  : {self.mau_xe}")
        print(f"Tốc độ  : {self.toc_do}")

Xe1 = XeHoi("Toyota", "Đỏ")
Xe1.chay()       # 50
Xe1.chay()       # 100

Xe2 = XeHoi("Honda", "Xanh")
Xe2.chay()       # 50 — hoàn toàn độc lập với Xe1

print("\n--- Thông tin xe 1 ---")
Xe1.thong_tin()
print("\n--- Thông tin xe 2 ---")
Xe2.thong_tin()

# Nhờ self.toc_do → Xe1 và Xe2 có tốc độ riêng, không ảnh hưởng nhau
````

### 1. Encapsulation — Đóng gói

> Gom **data** (attributes) + **behavior** (methods) vào 1 class.  
> Kiểm soát quyền truy cập từ bên ngoài — object tự quản lý state của mình.

**Tại sao cần đóng gói?** Không có đóng gói → ai cũng có thể ghi bất kỳ giá trị vào attribute, kể cả giá trị vô lý.

**Các mức độ truy cập:**

| Convention | Cách viết | Ý nghĩa |
|---|---|---|
| Public | `self.name` | Truy cập tự do từ bất kỳ đâu |
| Protected | `self._name` | Quy ước "nội bộ" — không nên đụng từ ngoài |
| Private | `self.__name` | Name-mangling → `_ClassName__name` — thực sự khó truy cập |

````python
class TaiKhoan:
    def __init__(self, so_du):
        self.so_du = so_du   # Public — ai cũng ghi được

tk = TaiKhoan(1_000_000)
print(tk.so_du)          # 1_000_000

tk.so_du = -999_999_999  # Python chấp nhận → Vô lý!
tk.so_du = "hahaha"      # Python chấp nhận → Vô lý!
print(tk.so_du)          # hahaha
````

````python
class ViDu:
    def __init__(self):
        self.ten    = "Nguyen An"      # Public    — ai cũng truy cập được
        self._tuoi  = 20               # Protected — chỉ dùng nội bộ (quy ước)
        self.__luong = 10_000_000      # Private   — không truy cập trực tiếp được

obj = ViDu()

print(obj.ten)             # ✅ Nguyen An
print(obj._tuoi)           # ⚠️ 20 — được nhưng không nên
# print(obj.__luong)       # ❌ AttributeError
print(obj._ViDu__luong)    # ⚠️ 10000000 — name-mangling trick
````

**Getter & Setter — kiểm soát đọc/ghi dữ liệu**

| | Getter | Setter |
|---|---|---|
| Mục đích | Đọc dữ liệu ra | Ghi dữ liệu vào (có kiểm tra) |
| Thủ công | `def get_x(self)` | `def set_x(self, val)` |
| Pythonic | `@property` | `@x.setter` |

> `@property` được ưu tiên vì interface không đổi — code bên ngoài vẫn dùng `obj.x` thay vì `obj.get_x()`.

```
hs.set_diem(-50)
      │
      ▼
 Setter kiểm tra
 -50 < 0 → SAI
      │
      ▼
 ❌ Raise ValueError  →  Dữ liệu được bảo vệ ✅
```

**Tại sao dùng `@property` thay vì getter/setter thủ công?** Khi thêm validation sau này, `@property` không làm vỡ code bên ngoài vì tên attribute vẫn giữ nguyên.

````python
# --- Lúc đầu: attribute thường ---
class SanPham:
    def __init__(self, gia):
        self.gia = gia

sp = SanPham(100_000)
print(sp.gia)     # code ngoài dùng: sp.gia
sp.gia = 200_000  # code ngoài dùng: sp.gia = ...


# --- Sau: thêm validation bằng @property (không làm vỡ code ngoài) ---
class SanPhamV2:
    def __init__(self, gia: float):
        self.__gia = gia

    @property
    def gia(self) -> float:
        return self.__gia

    @gia.setter
    def gia(self, g: float) -> None:
        if g < 0:
            raise ValueError("Gia khong the am!")
        self.__gia = g

sp2 = SanPhamV2(100_000)
print(sp2.gia)     # ✅ vẫn dùng sp2.gia — không cần sửa code ngoài
sp2.gia = 200_000  # ✅ vẫn dùng sp2.gia = ... — không cần sửa code ngoài
print(sp2.gia)
# sp2.gia = -100   # ❌ ValueError: Gia khong the am!
````

````python
class HocSinh:
    def __init__(self, diem: float):
        self.__diem = diem

    # --- Getter/Setter thủ công ---
    def get_diem(self) -> float:
        return self.__diem

    def set_diem(self, diem_moi: float) -> None:
        if not isinstance(diem_moi, (int, float)):
            raise TypeError("Diem phai la so!")
        if not (0 <= diem_moi <= 10):
            raise ValueError("Diem phai tu 0 den 10!")
        self.__diem = diem_moi

    # --- @property (Pythonic hơn) ---
    @property
    def diem(self) -> float:
        return self.__diem

    @diem.setter
    def diem(self, diem_moi: float) -> None:
        if not isinstance(diem_moi, (int, float)):
            raise TypeError("Diem phai la so!")
        if not (0 <= diem_moi <= 10):
            raise ValueError("Diem phai tu 0 den 10!")
        self.__diem = diem_moi


hs = HocSinh(8)

# Getter/Setter thủ công
print(hs.get_diem())   # 8
hs.set_diem(9)
print(hs.get_diem())   # 9

# @property — trông như biến thường
print(hs.diem)         # 9
hs.diem = 7
print(hs.diem)         # 7

# hs.diem = 100        # ❌ ValueError
# hs.diem = "abc"      # ❌ TypeError
````

````python
class NhanVien:
    def __init__(self, ten: str, luong: float):
        self.ten              = ten          # Public
        self._phong_ban       = "IT"         # Protected
        self.__luong          = luong        # Private
        self.__lich_su_luong  = []           # Private

    @property
    def luong(self) -> str:
        return f"{self.__luong:,} VND"

    @luong.setter
    def luong(self, luong_moi: float) -> None:
        if not isinstance(luong_moi, (int, float)):
            raise TypeError("Luong phai la so!")
        if luong_moi < 0:
            raise ValueError("Luong khong the am!")
        if luong_moi < self.__luong:
            raise ValueError("Khong the giam luong!")
        self.__lich_su_luong.append(f"{self.__luong:,} -> {luong_moi:,}")
        self.__luong = luong_moi

    @property
    def thue(self) -> str:       # computed property — chỉ đọc, không có setter
        return f"{self.__luong * 0.1:,} VND"

    def xem_lich_su(self) -> list:
        return self.__lich_su_luong or ["Chua co thay doi"]


nv = NhanVien("Nguyen An", 10_000_000)

print(nv.ten)          # ✅ Public — đọc trực tiếp
print(nv.luong)        # ✅ qua @property getter
print(nv.thue)         # ✅ computed — tự tính từ __luong

nv.luong = 12_000_000  # ✅ qua @property setter
nv.luong = 15_000_000  # ✅
print(nv.luong)
print(nv.xem_lich_su())

# nv.luong = -100       # ❌ ValueError
# nv.luong = 5_000_000  # ❌ ValueError — không thể giảm lương
# nv.luong = "abc"      # ❌ TypeError
````

### 2. Inheritance — Kế thừa

> Class con **(child)** thừa hưởng thuộc tính và phương thức từ class cha **(parent)**, giúp **tái sử dụng code** và **mở rộng chức năng**.

```
DongVat  ←  class cha (parent)
   ├── Cho   ←  class con: kế thừa + thêm/ghi đè
   └── Meo   ←  class con: kế thừa + thêm/ghi đè
```

````python
class DongVat:
    def __init__(self, ten: str, tuoi: int):
        self.ten  = ten
        self.tuoi = tuoi

    def gioi_thieu(self) -> None:
        print(f"Toi la {self.ten}, {self.tuoi} tuoi")

class Meo(DongVat):              # Meo kế thừa DongVat
    def keu(self) -> None:
        print(f"{self.ten}: Meo meo~")

meo = Meo("Mimi", 2)
meo.gioi_thieu()   # ✅ Kế thừa từ DongVat → "Toi la Mimi, 2 tuoi"
meo.keu()          # ✅ Method riêng      → "Mimi: Meo meo~"

print(isinstance(meo, DongVat))   # True — Meo là một DongVat
````

---
#### 2.1 Các Loại Kế Thừa

| Loại | Mô tả | Cú pháp |
|---|---|---|
| **Single** | 1 cha → 1 con | `class Cho(DongVat)` |
| **Multiple** | Nhiều cha → 1 con | `class Con(Cha, Me)` |
| **Multilevel** | Kế thừa theo chuỗi | `ChoCanhSat → Cho → DongVat` |

````python
# Single — đơn kế thừa (1 cha → 1 con)
class DongVat:
    def an(self):
        print("Dang an...")

class Cho(DongVat):
    def keu(self):
        print("Gau gau!")

cho = Cho()
cho.an()    # ✅ Kế thừa từ DongVat
cho.keu()   # ✅ Method riêng của Cho
````

````python
# Multiple — đa kế thừa (nhiều cha → 1 con)
class Cha:
    def ky_nang(self):
        print("Lap trinh")

class Me:
    def so_thich(self):
        print("Nau an")

class Con(Cha, Me):     # Con kế thừa cả Cha lẫn Me
    pass

con = Con()
con.ky_nang()   # ✅ Lap trinh
con.so_thich()  # ✅ Nau an
````

````python
# Multilevel — kế thừa theo chuỗi
class DongVat:
    def tho(self):
        print("Dang tho...")

class Cho(DongVat):
    def keu(self):
        print("Gau gau!")

class ChoCanhSat(Cho):   # ChoCanhSat → Cho → DongVat
    def nhiem_vu(self):
        print("Bao ve!")

cho_cs = ChoCanhSat()
cho_cs.tho()       # ✅ Từ DongVat
cho_cs.keu()       # ✅ Từ Cho
cho_cs.nhiem_vu()  # ✅ Của riêng ChoCanhSat
````

#### 2.2 Override — Ghi Đè Phương Thức

> Class con **định nghĩa lại** method cùng tên với class cha — Python luôn ưu tiên method của class con.

```
Python tìm method: Class con trước → Class cha sau
```

````python
class DongVat:
    def __init__(self, ten):
        self.ten = ten

    def keu(self):
        print(f"{self.ten}: ...")    # Mặc định

class Meo(DongVat):
    def keu(self):
        print(f"{self.ten}: Meo meo~")

class Cho(DongVat):
    def keu(self):
        print(f"{self.ten}: Gau gau!")

dongvat = DongVat("??")
meo     = Meo("Mimi")
cho     = Cho("Buddy")

dongvat.keu()   # ??: ...
meo.keu()       # Mimi: Meo meo~
cho.keu()       # Buddy: Gau gau!
````

#### 2.3 `super()` — Gọi Lại Method Của Cha

> Gọi lại phương thức của **class cha** từ class con, thay vì ghi đè hoàn toàn.

**Khi nào cần `super()`?**

| Tình huống | Cần `super()`? | Lý do |
|---|---|---|
| Class con **không có** `__init__` riêng | ❌ Không cần | Python tự kế thừa `__init__` của cha |
| Class con **có** `__init__` riêng (thêm attribute) | ✅ Cần | `__init__` cha bị ghi đè → phải gọi lại thủ công |
| Method thường muốn **giữ lại logic cha + thêm mới** | ✅ Cần | `super().method()` giữ logic cũ |

````python
class DongVat:
    def __init__(self, ten: str, tuoi: int):
        self.ten  = ten
        self.tuoi = tuoi

    def gioi_thieu(self) -> None:
        print(f"Toi la {self.ten}, {self.tuoi} tuoi")


# ✅ KHÔNG cần super() → Class con KHÔNG có __init__ riêng
class Meo(DongVat):
    def keu(self) -> None:
        print(f"{self.ten}: Meo meo~")
# Python tự dùng __init__ của DongVat

meo = Meo("Mimi", 2)   # ✅ Hoạt động bình thường
meo.gioi_thieu()


# ✅ CẦN super() → Class con CÓ __init__ riêng (thêm attribute mới)
class Cho(DongVat):
    def __init__(self, ten: str, tuoi: int, giong: str):
        super().__init__(ten, tuoi)  # Phục hồi logic của cha
        self.giong = giong           # Thêm attribute riêng

cho = Cho("Buddy", 3, "Labrador")   # ✅ Hoạt động bình thường
cho.gioi_thieu()
print(cho.giong)
````

#### 2.4 MRO — Thứ Tự Tìm Kiếm Phương Thức

> Python tìm method theo thứ tự: **Class con → Class cha → object**

Quan trọng khi dùng **đa kế thừa**: Python dùng thuật toán **C3 Linearization** đảm bảo mỗi class cha chỉ được gọi **1 lần duy nhất**, tránh xung đột.

Dùng `ClassName.__mro__` để xem thứ tự tìm kiếm.

````python
class A:
    def xin_chao(self):
        print("Xin chao tu A")

class B(A):
    def xin_chao(self):
        print("Xin chao tu B")
        super().xin_chao()

class C(A):
    def xin_chao(self):
        print("Xin chao tu C")
        super().xin_chao()

class D(B, C):
    def xin_chao(self):
        print("Xin chao tu D")
        super().xin_chao()

d = D()
d.xin_chao()
# Xin chao tu D
# Xin chao tu B   ← super() đi theo MRO, không gọi A trực tiếp
# Xin chao tu C   ← C được gọi trước A
# Xin chao tu A   ← A chỉ được gọi 1 lần duy nhất

print(D.__mro__)
# D → B → C → A → object
````

#### 3.2 Method Overriding — Ghi Đè Phương Thức

> Class con **ghi đè** method của class cha để có hành vi riêng

| Đặc điểm | Mô tả |
|---|---|
| Cần kế thừa | Class con extends class cha |
| Dùng `super()` | Gọi lại method class cha nếu cần |
| Phổ biến nhất | Được dùng rộng rãi trong OOP |

````python
class Shape:
    def area(self) -> float:
        return 0

    def describe(self) -> None:          # method dùng chung — tự gọi method bị override
        print(f"Dien tich: {self.area()}")

class Circle(Shape):
    def __init__(self, radius: float):
        self.radius = radius

    def area(self) -> float:             # Override
        return 3.14 * self.radius ** 2

class Rectangle(Shape):
    def __init__(self, width: float, height: float):
        self.width  = width
        self.height = height

    def area(self) -> float:             # Override
        return self.width * self.height

class Triangle(Shape):
    def __init__(self, base: float, height: float):
        self.base   = base
        self.height = height

    def area(self) -> float:             # Override
        return 0.5 * self.base * self.height

shapes = [Circle(5), Rectangle(4, 6), Triangle(3, 8)]
for shape in shapes:
    shape.describe()     # cùng gọi describe() → kết quả khác nhau
````

#### 3.3 Abstract Class

> **Bắt buộc** class con implement các method → tạo "hợp đồng" chung

| Đặc điểm | Mô tả |
|---|---|
| Không tạo được instance | Chỉ dùng làm class cha |
| `@abstractmethod` | Bắt buộc class con implement |
| An toàn nhất | Lỗi phát hiện sớm lúc định nghĩa class |

> **Khác Inheritance:** Inheritance chia sẻ code. **Abstraction** ép buộc class con tuân theo *contract*.

````python
from abc import ABC, abstractmethod

class Animal(ABC):
    def __init__(self, name: str):
        self.name = name

    @abstractmethod
    def speak(self) -> None: pass

    @abstractmethod
    def move(self) -> None: pass

    def info(self) -> None:     # method thường — class con kế thừa luôn
        print(f"Toi la {self.name}")
        self.speak()
        self.move()

class Bird(Animal):
    def speak(self) -> None: print(f"{self.name}: Tweet tweet!")
    def move(self)  -> None: print(f"{self.name} dang bay")

class Fish(Animal):
    def speak(self) -> None: print(f"{self.name}: ... (im lang)")
    def move(self)  -> None: print(f"{self.name} dang boi")

# Animal("test")   # ❌ TypeError — không tạo được object từ abstract class

Bird("Chim se").info()
print("---")
Fish("Ca vang").info()
````

#### 3.4 Operator Overloading — Nạp Chồng Toán Tử

> Tùy chỉnh hành vi của các **toán tử** (`+`, `*`, `==`, ...) cho class của mình

| Magic Method | Toán tử / Hàm |
|---|---|
| `__add__` | `+` |
| `__sub__` | `-` |
| `__mul__` | `*` |
| `__eq__` | `==` |
| `__lt__` | `<` |
| `__str__` | `str()` |
| `__len__` | `len()` |

````python
class Vector:
    def __init__(self, x: float, y: float):
        self.x = x
        self.y = y

    def __add__(self, other: "Vector") -> "Vector":   # Nạp chồng +
        return Vector(self.x + other.x, self.y + other.y)

    def __mul__(self, scalar: float) -> "Vector":     # Nạp chồng *
        return Vector(self.x * scalar, self.y * scalar)

    def __eq__(self, other: "Vector") -> bool:        # Nạp chồng ==
        return self.x == other.x and self.y == other.y

    def __str__(self) -> str:                         # Nạp chồng str()
        return f"Vector({self.x}, {self.y})"

    def __len__(self) -> int:                         # Nạp chồng len()
        return int((self.x**2 + self.y**2) ** 0.5)

v1 = Vector(1, 2)
v2 = Vector(3, 4)
# Khi gọi toán tử "+", Python sẽ tìm kiếm method __add__ trong class Vector và dịch thành v1.__add__(v2)
# print() chỉ hiển thị được CHUỖI (string)
# → Khi nhận vào một object, nó phải chuyển object → chuỗi
# → Python tự động gọi __str__() để lấy chuỗi
# Nếu không có __str__, Python sẽ in ra object
print(v1 + v2)     # Vector(4, 6)

print(v1 * 3)      # Vector(3, 6)

print(v1 == v2)    # False

print(len(v2))     # 5
````

#### 3.5 Method Overloading — Nạp Chồng Phương Thức

> Python **không hỗ trợ trực tiếp** → mô phỏng bằng `*args` hoặc **default parameters**

````python
class Calculator:
    def add(self, *args) -> float:             # *args → nhận số lượng đối số bất kỳ
        return sum(args)

    def multiply(self, a, b=1, c=1) -> float:  # default params → mô phỏng overloading
        return a * b * c

calc = Calculator()

print(calc.add(1, 2))         # 3
print(calc.add(1, 2, 3))      # 6
print(calc.add(1, 2, 3, 4))   # 10

print(calc.multiply(5))        # 5
print(calc.multiply(5, 2))     # 10
print(calc.multiply(5, 2, 3))  # 30
````

#### Ví Dụ Thực Tế — Hệ Thống Thanh Toán

> Kết hợp **Abstract Class + Method Overriding + Duck Typing**

````python
from abc import ABC, abstractmethod

class PaymentMethod(ABC):
    @abstractmethod
    def pay(self, amount: float) -> bool: pass

    @abstractmethod
    def refund(self, amount: float) -> bool: pass

class CreditCard(PaymentMethod):
    def __init__(self, card_number: str):
        self.card_number = card_number[-4:]

    def pay(self, amount: float) -> bool:
        print(f"Thanh toan {amount:,.0f}d bang the ****{self.card_number}")
        return True

    def refund(self, amount: float) -> bool:
        print(f"Hoan tien {amount:,.0f}d ve the ****{self.card_number}")
        return True

class MoMo(PaymentMethod):
    def __init__(self, phone: str):
        self.phone = phone

    def pay(self, amount: float) -> bool:
        print(f"Thanh toan {amount:,.0f}d qua MoMo ({self.phone})")
        return True

    def refund(self, amount: float) -> bool:
        print(f"Hoan {amount:,.0f}d ve vi MoMo ({self.phone})")
        return True

class Cash(PaymentMethod):
    def pay(self, amount: float) -> bool:
        print(f"Thanh toan tien mat: {amount:,.0f}d")
        return True

    def refund(self, amount: float) -> bool:
        print(f"Tra lai tien mat: {amount:,.0f}d")
        return True

class OrderProcessor:
    def __init__(self, payment: PaymentMethod):
        self.payment = payment

    def checkout(self, amount: float) -> None:
        print(f"\nDon hang: {amount:,.0f}d")
        if self.payment.pay(amount):
            print("Thanh toan thanh cong!")

orders = [
    OrderProcessor(CreditCard("1234567890123456")),
    OrderProcessor(MoMo("0901234567")),
    OrderProcessor(Cash()),
]

for order in orders:
    order.checkout(150_000)
````

### 4. Abstraction — Trừu tượng

> **Tính trừu tượng** = Ẩn **chi tiết triển khai (HOW)**, chỉ lộ **giao diện cần thiết (WHAT)**

```
┌──────────────────────────────────────────────────┐
│                TÍNH TRỪU TƯỢNG                   │
│                                                  │
│   Người dùng thấy       │   Được ẩn đi           │
│   ┌─────────────────┐   │   ┌─────────────────┐  │
│   │  Giao diện      │   │   │  Cách thực hiện │  │
│   │  Tên phương     │   │   │  Logic nội bộ   │  │
│   │  thức cần gọi   │   │   │  Thuật toán     │  │
│   │  (WHAT)         │   │   │  (HOW)          │  │
│   └─────────────────┘   │   └─────────────────┘  │
└──────────────────────────────────────────────────┘
```

**Công cụ triển khai — module `abc`:**

| Thành phần | Ý nghĩa |
|---|---|
| `ABC` | Abstract Base Class — class cha trừu tượng |
| `@abstractmethod` | Method **bắt buộc** override ở class con |
| `@property` + `@abstractmethod` | Property bắt buộc override |

> **Khác Inheritance:** Inheritance chia sẻ code.  
> **Abstraction** ép buộc class con tuân theo *contract* — đảm bảo mọi class con đều có đủ interface.

````python
from abc import ABC, abstractmethod

# BƯỚC 1: Định nghĩa lớp trừu tượng
class HinhHoc(ABC):
    @abstractmethod
    def tinh_dien_tich(self): pass

    @abstractmethod
    def tinh_chu_vi(self): pass

    def mo_ta(self):     # method thường — class con kế thừa luôn
        return (f"Dien tich : {self.tinh_dien_tich():.2f}\n"
                f"Chu vi    : {self.tinh_chu_vi():.2f}")

# BƯỚC 2: Class con bắt buộc implement đủ
class HinhTron(HinhHoc):
    def __init__(self, ban_kinh: float):
        self.ban_kinh = ban_kinh

    def tinh_dien_tich(self) -> float:    # ✅ implement
        return 3.14159 * self.ban_kinh ** 2

    def tinh_chu_vi(self) -> float:       # ✅ implement
        return 2 * 3.14159 * self.ban_kinh

class HinhChuNhat(HinhHoc):
    def __init__(self, dai: float, rong: float):
        self.dai  = dai
        self.rong = rong

    def tinh_dien_tich(self) -> float:    # ✅ implement
        return self.dai * self.rong

    def tinh_chu_vi(self) -> float:       # ✅ implement
        return 2 * (self.dai + self.rong)

# HinhHoc()   # ❌ TypeError — không tạo được object từ abstract class
tron     = HinhTron(5)
chu_nhat = HinhChuNhat(4, 6)

print("=== Hinh Tron ===")
print(tron.mo_ta())

print("\n=== Hinh Chu Nhat ===")
print(chu_nhat.mo_ta())
````

#### 4.1 Abstract Method — Phương Thức Trừu Tượng

> Method có `@abstractmethod` **bắt buộc** class con override.  
> Class con không implement đủ → `TypeError` ngay khi khởi tạo object.

````python
from abc import ABC, abstractmethod

class DongVat(ABC):
    @abstractmethod             # ← phương thức trừu tượng
    def keu(self): pass         # class con luôn bắt buộc phải có hàm keu()

    def gioi_thieu(self):       # ← phương thức bình thường
        return f"Toi keu: {self.keu()}"


class Cho(DongVat):
    def keu(self): return "Gau Gau!"   # ✅ implement

class Meo(DongVat):
    def keu(self): return "Meo Meo!"   # ✅ implement


cho = Cho()
meo = Meo()
print(cho.gioi_thieu())   # Toi keu: Gau Gau!
print(meo.gioi_thieu())   # Toi keu: Meo Meo!

# DongVat()   # ❌ TypeError — không implement keu()
````

#### 4.2 Abstract Property — Thuộc Tính Trừu Tượng

> Kết hợp `@property` + `@abstractmethod` để bắt buộc class con định nghĩa thuộc tính

Thứ tự decorator: `@property` **trước**, `@abstractmethod` **sau** (ngay trên `def`).

````python
from abc import ABC, abstractmethod

class DongVat(ABC):
    @property
    @abstractmethod             # đặt DƯỚI @property
    def ten_loai(self): pass

    @abstractmethod
    def keu(self): pass

    def gioi_thieu(self):
        return f"Toi la {self.ten_loai}, toi keu: {self.keu()}"


class Cho(DongVat):
    @property
    def ten_loai(self): return "Con Cho"   # ✅ override property

    def keu(self): return "Gau Gau!"


class Meo(DongVat):
    @property
    def ten_loai(self): return "Con Meo"

    def keu(self): return "Meo Meo!"


for dv in [Cho(), Meo()]:
    print(dv.gioi_thieu())
# Toi la Con Cho, toi keu: Gau Gau!
# Toi la Con Meo, toi keu: Meo Meo!
````

#### 4.3 Template Method — Phương Thức Mẫu

> Định nghĩa **khung sườn cố định** trong lớp cha, lớp con chỉ điền vào **bước biến đổi**

```
Abstract Class
├── @abstractmethod buoc_1()     ← lớp con tự implement
├── @abstractmethod buoc_2()     ← lớp con tự implement
└── khung_suon_co_dinh()         ← lớp cha kiểm soát thứ tự gọi
```

````python
from abc import ABC, abstractmethod

class PhuongThucThanhToan(ABC):
    # Các bước biến đổi — lớp con tự implement
    @abstractmethod
    def ket_noi(self) -> bool: pass

    @abstractmethod
    def thanh_toan(self, so_tien: float) -> bool: pass

    @abstractmethod
    def hoan_tien(self, so_tien: float) -> bool: pass

    # Khung sườn cố định — lớp con KHÔNG được đổi
    def xu_ly_giao_dich(self, so_tien: float):
        print(f"Bat dau giao dich: {so_tien:,}d")
        if not self.ket_noi():
            print("Ket noi that bai!")
            return False
        if self.thanh_toan(so_tien):
            print(f"Thanh toan {so_tien:,}d thanh cong!")
            return True
        print("Thanh toan that bai!")
        return False
````

#### Ví Dụ Thực Tế — Hệ Thống Thanh Toán

> Kết hợp **Abstract Class + Template Method + Encapsulation**  
> `GioHang` (client) **không cần biết** Visa hay MoMo xử lý thế nào — chỉ gọi `xu_ly_giao_dich()`

````python
from abc import ABC, abstractmethod

class PhuongThucThanhToan(ABC):
    @abstractmethod
    def ket_noi(self) -> bool: pass

    @abstractmethod
    def thanh_toan(self, so_tien: float) -> bool: pass

    @abstractmethod
    def hoan_tien(self, so_tien: float) -> bool: pass

    def xu_ly_giao_dich(self, so_tien: float):       # Template Method — khung sườn cố định
        print(f"Bat dau giao dich: {so_tien:,}d")
        if not self.ket_noi():
            print("Ket noi that bai!")
            return False
        if self.thanh_toan(so_tien):
            print(f"Thanh toan {so_tien:,}d thanh cong!")
            return True
        print("Thanh toan that bai!")
        return False


class ThanhToanVisa(PhuongThucThanhToan):
    def __init__(self, so_the: str, ma_bao_mat: str):
        self.__so_the     = so_the          # 🔒 ẩn thông tin nhạy cảm
        self.__ma_bao_mat = ma_bao_mat

    def ket_noi(self) -> bool:
        print("  Ket noi Visa Gateway...")
        return True

    def thanh_toan(self, so_tien: float) -> bool:
        print(f"  Xu ly the Visa *{self.__so_the[-4:]}...")
        return True

    def hoan_tien(self, so_tien: float) -> bool:
        print(f"  Hoan {so_tien:,}d ve the Visa...")
        return True


class ThanhToanMomo(PhuongThucThanhToan):
    def __init__(self, so_dien_thoai: str):
        self.__sdt = so_dien_thoai          # 🔒 ẩn số điện thoại

    def ket_noi(self) -> bool:
        print("  Ket noi MoMo API...")
        return True

    def thanh_toan(self, so_tien: float) -> bool:
        an = f"{self.__sdt[:3]}****{self.__sdt[-3:]}"
        print(f"  Xu ly MoMo {an}...")
        return True

    def hoan_tien(self, so_tien: float) -> bool:
        print(f"  Hoan {so_tien:,}d ve vi MoMo...")
        return True


class GioHang:
    def __init__(self, tong_tien: float):
        self.tong_tien = tong_tien

    def checkout(self, phuong_thuc: PhuongThucThanhToan):
        return phuong_thuc.xu_ly_giao_dich(self.tong_tien)  # không cần biết Visa hay MoMo


gio_hang = GioHang(500_000)

print("=== Thanh toan Visa ===")
gio_hang.checkout(ThanhToanVisa("1234567890123456", "123"))

print("\n=== Thanh toan MoMo ===")
gio_hang.checkout(ThanhToanMomo("0901234567"))
````

#### Kiểm Tra & Xử Lý Lỗi

> Class con chưa implement đủ `@abstractmethod` → `TypeError` ngay khi khởi tạo object.  
> Dùng `ClassName.__abstractmethods__` để xem danh sách method còn thiếu.

````python
from abc import ABC, abstractmethod

class KiemTra(ABC):
    @abstractmethod
    def phuong_thuc(self): pass

class ConChua(KiemTra):
    pass                        # ❌ chưa implement

class ConDayDu(KiemTra):
    def phuong_thuc(self):
        return "Da implement!"

# Kiểm tra abstract methods
print(KiemTra.__abstractmethods__)   # frozenset({'phuong_thuc'})

# Lỗi khi chưa implement đủ
try:
    obj = ConChua()
except TypeError as e:
    print(f"Loi: {e}")

# Thành công khi implement đủ
obj2 = ConDayDu()
print(obj2.phuong_thuc())            # Da implement!

# Kiểm tra quan hệ kế thừa
print(isinstance(obj2, KiemTra))     # True
print(isinstance(obj2, ABC))         # True
````
