# Two Pointers (Kỹ thuật Hai Con Trỏ)


## 1. Định nghĩa

**Two Pointers** là kỹ thuật sử dụng **hai biến (con trỏ)** để cùng duyệt qua một cấu trúc dữ liệu như mảng hoặc chuỗi.

Các con trỏ này có thể:
- Bắt đầu ở **hai đầu đối diện** và di chuyển vào giữa
- Bắt đầu ở **cùng một đầu** và di chuyển cùng chiều với tốc độ khác nhau

### Mục tiêu chính
| Tiêu chí | Brute Force | Two Pointers |
|---|---|---|
| Time Complexity | O(n²) | **O(n)** |
| Space Complexity | O(1) ~ O(n) | **O(1)** |

## 2. Hai mô hình di chuyển chính

### Mô hình 1: Từ hai phía vào giữa (Opposite Direction)
```
[1, 3, 5, 7, 9, 11]
 ^                ^
left             right
  --> di chuyển vào giữa <--
```
- Áp dụng cho **mảng đã sắp xếp**
- Một con trỏ ở đầu (`left`), một ở cuối (`right`)
- Di chuyển dần vào trung tâm cho đến khi `left >= right`

---

### Mô hình 2: Cùng chiều (Same Direction / Slow-Fast)
```
[1, 1, 2, 3, 3, 4]
 ^  ^
slow fast
  --> cả hai di chuyển sang phải -->
```
- `slow`: ghi nhớ vị trí phần tử hợp lệ
- `fast`: duyệt toàn bộ mảng
- Thường dùng trong **Sliding Window**, loại bỏ trùng lặp

## 3. Khi nào nên dùng Two Pointers?

✅ Dữ liệu **có thứ tự** (sorted array/string)  
✅ Tìm kiếm **cặp/bộ ba** phần tử thỏa điều kiện  
✅ Xử lý **đoạn con** (subarray / substring)  
✅ **Loại bỏ trùng lặp** in-place  
✅ Bài toán **palindrome**

## 4. Ví dụ thực tế

### Ví dụ 1: Tìm cặp số có tổng bằng target

````python
def find_pair_with_sum(arr: list[int], target: int) -> bool:
    """Tìm cặp số trong mảng đã sắp xếp có tổng bằng target.

    Args:
        arr: Mảng số nguyên đã sắp xếp tăng dần.
        target: Giá trị tổng cần tìm.

    Returns:
        True nếu tìm thấy cặp, False nếu không.
    """
    left, right = 0, len(arr) - 1

    while left < right:
        current_sum = arr[left] + arr[right]

        if current_sum == target:
            print(f"Found pair: arr[{left}]={arr[left]} + arr[{right}]={arr[right]} = {target}")
            return True
        elif current_sum < target:
            left += 1   # tổng nhỏ hơn → tăng left để tăng tổng
        else:
            right -= 1  # tổng lớn hơn → giảm right để giảm tổng

    return False


# Test
arr = [1, 3, 5, 7, 9, 11]
print(find_pair_with_sum(arr, 10))  # True: 1+9 hoặc 3+7
print(find_pair_with_sum(arr, 20)) # False
````

````python
a = [1, 3, 5, 6, 7, 9, 10, 12]

# tìm các cặp số có tổng bằng 10
def find_pair(arr: list[int], target: int) -> list[tuple[int, int]]:
    left = 0
    right = len(arr) - 1
    result = []

    while left < right:
        total = arr[left] + arr[right]
        if total == target:
            print(f"Found pair: arr[{left}]={arr[left]} + arr[{right}]={arr[right]} = {target}")
            result.append((arr[left], arr[right]))
            left += 1   # tiếp tục tìm cặp tiếp theo
            right -= 1
        elif total < target:
            left += 1
        else:
            right -= 1

    return result

print(find_pair(a, 10))
````

### Ví dụ 2: Kiểm tra Palindrome

````python
def is_palindrome(s: str) -> bool:
    """Kiểm tra chuỗi có phải palindrome không (bỏ qua ký tự đặc biệt).

    Args:
        s: Chuỗi đầu vào.

    Returns:
        True nếu là palindrome.
    """
    # Chuẩn hóa: chỉ giữ chữ và số, viết thường
    cleaned = [c.lower() for c in s if c.isalnum()]

    left, right = 0, len(cleaned) - 1

    while left < right:
        if cleaned[left] != cleaned[right]:
            return False
        left += 1
        right -= 1

    return True


# Test
print(is_palindrome("A man, a plan, a canal: Panama"))  # True
print(is_palindrome("race a car"))                       # False
print(is_palindrome("racecar"))                          # True
````

### Ví dụ 3: Loại bỏ phần tử trùng lặp trong mảng đã sắp xếp (In-place)

````python
def remove_duplicates(nums: list[int]) -> int:
    """Loại bỏ trùng lặp in-place trên mảng đã sắp xếp.

    Args:
        nums: Mảng số nguyên đã sắp xếp (thay đổi trực tiếp).

    Returns:
        Số phần tử duy nhất.
    """
    if not nums:
        return 0

    slow = 0  # slow trỏ tới vị trí phần tử duy nhất tiếp theo

    for fast in range(1, len(nums)):
        if nums[fast] != nums[slow]:  # tìm thấy phần tử mới
            slow += 1
            nums[slow] = nums[fast]   # ghi vào vị trí slow

    return slow + 1  # số phần tử duy nhất


# Test
nums = [1, 1, 2, 3, 3, 4, 4, 5]
k = remove_duplicates(nums)
print(f"Số phần tử duy nhất: {k}")
print(f"Mảng sau xử lý: {nums[:k]}")  # [1, 2, 3, 4, 5]
````

### Ví dụ 4: Container With Most Water

Cho mảng chiều cao `height`, tìm hai đường thẳng tạo thành container chứa nhiều nước nhất.

**Chiến lược**: Diện tích = `min(height[left], height[right]) × (right - left)`. Luôn di chuyển con trỏ có chiều cao **nhỏ hơn**.

````python
def max_area(height: list[int]) -> int:
    """Tìm diện tích container chứa nhiều nước nhất.

    Args:
        height: Danh sách chiều cao các đường thẳng.

    Returns:
        Diện tích lớn nhất có thể.
    """
    left, right = 0, len(height) - 1
    max_water = 0

    while left < right:
        width = right - left
        water = min(height[left], height[right]) * width
        max_water = max(max_water, water)

        # Di chuyển con trỏ có chiều cao nhỏ hơn (cố định chiều cao lớn)
        if height[left] < height[right]:
            left += 1
        else:
            right -= 1

    return max_water


# Test
height = [1, 8, 6, 2, 5, 4, 8, 3, 7]
print(f"Diện tích lớn nhất: {max_area(height)}")  # 49
````

| Bài toán | Time | Space | Ghi chú |
|---|---|---|---|
| Tìm cặp tổng bằng target | O(n) | O(1) | Mảng phải sorted |
| Kiểm tra palindrome | O(n) | O(n) | O(n) do bước clean chuỗi |
| Loại bỏ trùng lặp | O(n) | O(1) | In-place |
| Container with most water | O(n) | O(1) | |

> So sánh với Brute Force (nested loop): O(n²) time → **Two Pointers tiết kiệm đáng kể**

## 7. Anti-patterns cần tránh

| Lỗi thường gặp | Cách tránh |
|---|---|
| Dùng Two Pointers trên mảng **chưa sắp xếp** (Opposite Direction) | Sort trước hoặc dùng mô hình khác |
| `left <= right` thay vì `left < right` | Gây vòng lặp vô tận / so sánh phần tử với chính nó |
| Không cập nhật cả hai con trỏ đúng hướng | Vẽ trace tay trước khi code |
| Quên xử lý mảng rỗng hoặc một phần tử | Thêm guard clause ở đầu hàm |

## 9. Bài tập: 3Sum

Tìm tất cả bộ ba phần tử trong mảng có tổng bằng 0 (không trùng lặp).

````python
def three_sum(nums: list[int]) -> list[list[int]]:
    """Tìm tất cả bộ ba số có tổng bằng 0.

    Args:
        nums: Danh sách số nguyên.

    Returns:
        Danh sách các bộ ba không trùng lặp.
    """
    nums.sort()
    result = []

    for i in range(len(nums) - 2):
        # Bỏ qua phần tử trùng ở vị trí i
        if i > 0 and nums[i] == nums[i - 1]:
            continue

        left, right = i + 1, len(nums) - 1

        while left < right:
            total = nums[i] + nums[left] + nums[right]

            if total == 0:
                result.append([nums[i], nums[left], nums[right]])
                # Bỏ qua trùng lặp ở left và right
                while left < right and nums[left] == nums[left + 1]:
                    left += 1
                while left < right and nums[right] == nums[right - 1]:
                    right -= 1
                left += 1
                right -= 1
            elif total < 0:
                left += 1
            else:
                right -= 1

    return result


# Test
print(three_sum([-1, 0, 1, 2, -1, -4]))  # [[-1, -1, 2], [-1, 0, 1]]
print(three_sum([0, 0, 0]))              # [[0, 0, 0]]
````
