# Bubble Sort

## 1. Định nghĩa

**Bubble Sort** (Sắp xếp nổi bọt) là một thuật toán sắp xếp cơ bản, hoạt động bằng cách **liên tục so sánh các phần tử kề nhau** và hoán đổi chúng nếu chúng đang ở sai thứ tự.

Tên gọi xuất phát từ cách các phần tử lớn hơn "nổi lên" về đúng vị trí của chúng — giống như bong bóng khí nổi lên trong nước.

## 2. Cách hoạt động

Thuật toán thực hiện nhiều **lượt duyệt (pass)** qua dữ liệu:

1. Trong mỗi lượt, so sánh từng cặp phần tử kề nhau.
2. Nếu cặp đó sai thứ tự → **hoán đổi**.
3. Sau mỗi lượt, phần tử **lớn nhất chưa sắp xếp** sẽ "nổi" về đúng vị trí cuối.
4. Lặp lại cho đến khi **không có hoán đổi nào** xảy ra trong một lượt hoàn chỉnh.

### Minh họa trực quan

```
Mảng ban đầu: [5, 3, 8, 4, 2]

Lượt 1:
  [5, 3, 8, 4, 2] → so sánh 5 và 3 → swap → [3, 5, 8, 4, 2]
  [3, 5, 8, 4, 2] → so sánh 5 và 8 → ok    → [3, 5, 8, 4, 2]
  [3, 5, 8, 4, 2] → so sánh 8 và 4 → swap → [3, 5, 4, 8, 2]
  [3, 5, 4, 8, 2] → so sánh 8 và 2 → swap → [3, 5, 4, 2, 8]  ✓ (8 đúng vị trí)

Lượt 2:
  [3, 5, 4, 2, 8] → ... → [3, 4, 2, 5, 8]  ✓ (8, 5 đúng vị trí)

... tiếp tục cho đến khi mảng được sắp xếp.
```

## 3. Các bước thuật toán

```
1. Lấy độ dài mảng n.
2. Vòng lặp ngoài i từ 0 đến n-1 (kiểm soát số lượt duyệt).
3. Vòng lặp trong j từ 0 đến n-i-2 (phần chưa sắp xếp).
4. Nếu arr[j] > arr[j+1] → hoán đổi arr[j] và arr[j+1], đánh dấu swapped = True.
5. Nếu sau một lượt không có hoán đổi → dừng sớm (tối ưu).
```

## 4. Cài đặt Python

````python
def bubble_sort(arr: list) -> list:
    """Sắp xếp danh sách theo thứ tự tăng dần bằng Bubble Sort.

    Args:
        arr: Danh sách các phần tử có thể so sánh được.

    Returns:
        Danh sách đã được sắp xếp (in-place).
    """
    n = len(arr)

    for i in range(n):
        swapped = False  # Cờ tối ưu: dừng sớm nếu không có hoán đổi

        # Chỉ duyệt phần chưa sắp xếp (n-i-1 phần tử)
        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]  # Hoán đổi kiểu Python
                swapped = True

        if not swapped:  # Mảng đã sắp xếp → thoát sớm
            break

    return arr
````

````python
# --- Chạy thử ---
data = [64, 34, 25, 12, 22, 11, 90]
print("Trước khi sắp xếp:", data)

sorted_data = bubble_sort(data.copy())  # .copy() để giữ nguyên mảng gốc
print("Sau khi sắp xếp: ", sorted_data)
````

````python
# --- Test thêm các trường hợp đặc biệt ---
print(bubble_sort([]))           # Mảng rỗng → []
print(bubble_sort([1]))          # Một phần tử → [1]
print(bubble_sort([3, 1, 2]))    # Mảng ngẫu nhiên → [1, 2, 3]
print(bubble_sort([1, 2, 3]))    # Đã sắp xếp → [1, 2, 3] (O(n) nhờ early termination)
print(bubble_sort([3, 2, 1]))    # Ngược chiều (worst case) → [1, 2, 3]
````

## 5. Phân tích độ phức tạp

| Trường hợp | Độ phức tạp thời gian | Ghi chú |
|---|---|---|
| **Tốt nhất** (Best) | O(n) | Mảng đã sắp xếp → early termination dừng ngay lượt 1 |
| **Trung bình** (Average) | O(n²) | Mảng ngẫu nhiên |
| **Tệ nhất** (Worst) | O(n²) | Mảng sắp xếp ngược chiều |
| **Không gian** (Space) | O(1) | Sắp xếp tại chỗ (in-place), không cần bộ nhớ thêm |

## 6. Ưu điểm & Nhược điểm

### Ưu điểm
- **Đơn giản, dễ hiểu** — logic trực quan, phù hợp học thuật toán.
- **In-place** — không cần bộ nhớ phụ.
- **Stable sort** — giữ nguyên thứ tự tương đối của các phần tử bằng nhau.
- **Tốt cho dữ liệu gần sắp xếp** — nhờ cơ chế early termination.

### Nhược điểm
- **Hiệu suất kém** trên tập dữ liệu lớn — O(n²).
- **Không thực dụng trong production** — thua xa Merge Sort O(n log n) hay Tim Sort.
- Số lượng hoán đổi cao → tốn chi phí ghi bộ nhớ.

## 7. So sánh với các thuật toán khác

| Thuật toán | Best | Average | Worst | Space | Stable? |
|---|---|---|---|---|---|
| **Bubble Sort** | O(n) | O(n²) | O(n²) | O(1) | ✅ |
| Selection Sort | O(n²) | O(n²) | O(n²) | O(1) | ❌ |
| Insertion Sort | O(n) | O(n²) | O(n²) | O(1) | ✅ |
| Merge Sort | O(n log n) | O(n log n) | O(n log n) | O(n) | ✅ |
| Quick Sort | O(n log n) | O(n log n) | O(n²) | O(log n) | ❌ |
| Tim Sort (Python built-in) | O(n) | O(n log n) | O(n log n) | O(n) | ✅ |

## 8. Khi nào nên dùng?

- **Học thuật toán** — hiểu khái niệm so sánh & hoán đổi.
- **Dữ liệu rất nhỏ** (< 20 phần tử) — chi phí overhead của các thuật toán phức tạp hơn không đáng.
- **Dữ liệu gần như đã sắp xếp** — early termination giúp tiếp cận O(n).
- ❌ **Không dùng** cho dữ liệu lớn trong môi trường production — hãy dùng `sorted()` hoặc `.sort()` của Python (Tim Sort).
