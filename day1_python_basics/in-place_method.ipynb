# In-place Methods

**In-place method** là method biến đổi object trực tiếp tại chỗ — **không tạo object mới** và **luôn return `None`**.

> Python quy tắc nhất quán: nếu method thay đổi object tại chỗ → return `None`, để tránh nhầm lẫn với object mới.

## `list.reverse()`

- Đảo ngược thứ tự list **tại chỗ**
- Return `None`
- Muốn giữ list gốc → dùng `reversed()` hoặc `[::-1]`

```python
ds = [1, 2, 3, 4, 5]
ket_qua = ds.reverse()

print(ds)        # [5, 4, 3, 2, 1] — list gốc đã thay đổi
print(ket_qua)   # None
```

```python
# Muốn gán kết quả vào biến mới → dùng reversed() hoặc slicing
ds = [1, 2, 3, 4, 5]

dao1 = list(reversed(ds))  # tạo list mới
dao2 = ds[::-1]            # tạo list mới bằng slicing

print(dao1)  # [5, 4, 3, 2, 1]
print(dao2)  # [5, 4, 3, 2, 1]
print(ds)    # [1, 2, 3, 4, 5] — list gốc không đổi
```

## `list.sort()`

- Sắp xếp list **tại chỗ**
- Return `None`
- Tham số: `reverse=True` để đảo chiều, `key=func` để tùy chỉnh tiêu chí sắp xếp
- Muốn giữ list gốc → dùng `sorted()`

```python
so = [3, 1, 4, 1, 5, 9, 2]
ket_qua = so.sort()

print(so)        # [1, 1, 2, 3, 4, 5, 9] — list gốc đã thay đổi
print(ket_qua)   # None
```

```python
# reverse=True
so = [3, 1, 4, 1, 5, 9, 2]
so.sort(reverse=True)
print(so)  # [9, 5, 4, 3, 2, 1, 1]
```

```python
# Sắp xếp theo key = func
ten = ["Chuối", "táo", "Xoài", "nho"]
ten.sort(key=str.lower)   # sắp xếp không phân biệt hoa thường
print(ten)  # ['Chuối', 'nho', 'táo', 'Xoài']
# Sắp xếp theo nhiều tiêu chí
ten = ["Chuối", "táo", "Xoài", "nho"]
ten.sort(key=lambda x: (len(x), x.lower())) # Thứ tự sắp xếp sẽ ưu tiên theo len() --> lower()
print(ten)
```

```python
# Muốn giữ list gốc → dùng sorted()
so = [3, 1, 4, 1, 5]
sap_xep = sorted(so)       # tạo list mới

print(sap_xep)  # [1, 1, 3, 4, 5]
print(so)       # [3, 1, 4, 1, 5] — list gốc không đổi
```

## `list.append()`

- Thêm 1 phần tử vào **cuối** list tại chỗ
- Return `None`

```python
ds = [1, 2, 3]
ket_qua = ds.append(4)

print(ds)        # [1, 2, 3, 4]
print(ket_qua)   # None

# Lỗi hay gặp: gán kết quả append vào chính biến đó
ds = [1, 2, 3]
ds = ds.append(4)   # ds bây giờ là None!
print(ds)           # None
```

## `list.extend()`

- Nối thêm nhiều phần tử từ một iterable vào cuối list tại chỗ
- Return `None`
- Khác `append()`: `extend([4,5])` thêm từng phần tử; `append([4,5])` thêm nguyên cả list như 1 phần tử

```python
ds = [1, 2, 3]
ket_qua = ds.extend([4, 5])

print(ds)        # [1, 2, 3, 4, 5]
print(ket_qua)   # None

# So sánh extend vs append
a = [1, 2, 3]
b = [1, 2, 3]
a.extend([4, 5])   # [1, 2, 3, 4, 5]
b.append([4, 5])   # [1, 2, 3, [4, 5]]
print(a)
print(b)
```

## `list.insert()`

- Chèn 1 phần tử tại vị trí `i` tại chỗ
- Return `None`

```python
ds = ["táo", "chuối", "cam"]
ket_qua = ds.insert(1, "nho")

print(ds)        # ['táo', 'nho', 'chuối', 'cam']
print(ket_qua)   # None
```

## `list.remove()`

- Xóa **lần xuất hiện đầu tiên** của giá trị tại chỗ
- Return `None`
- Báo `ValueError` nếu giá trị không tồn tại

```python
ds = ["táo", "chuối", "táo", "cam"]
ket_qua = ds.remove("táo")

print(ds)        # ['chuối', 'táo', 'cam'] — chỉ xóa lần đầu tiên
print(ket_qua)   # None

# An toàn khi không chắc phần tử có tồn tại
if "xoài" in ds:
    ds.remove("xoài")
```

## `list.clear()`

- Xóa **toàn bộ** phần tử tại chỗ
- Return `None`
- Khác `ds = []`: `clear()` giữ nguyên object, `ds = []` tạo object mới

```python
ds = [1, 2, 3]
ket_qua = ds.clear()

print(ds)        # []
print(ket_qua)   # None

# Sự khác biệt khi có nhiều biến cùng trỏ vào 1 list
a = [1, 2, 3]
b = a           # b và a cùng trỏ vào 1 object

a.clear()       # xóa nội dung object gốc
print(b)        # [] — b cũng bị ảnh hưởng

a = [1, 2, 3]
b = a
a = []          # a trỏ sang object mới
print(b)        # [1, 2, 3] — b không bị ảnh hưởng
```

## Bảng Tổng Hợp

| Method | Tác dụng | In-place | Return | Thay thế nếu cần giữ gốc |
|---|---|---|---|---|
| `list.reverse()` | Đảo ngược | ✅ | `None` | `reversed()` hoặc `[::-1]` |
| `list.sort()` | Sắp xếp | ✅ | `None` | `sorted()` |
| `list.append(x)` | Thêm 1 phần tử vào cuối | ✅ | `None` | `ds + [x]` |
| `list.extend(it)` | Nối nhiều phần tử vào cuối | ✅ | `None` | `ds + list(it)` |
| `list.insert(i, x)` | Chèn tại vị trí `i` | ✅ | `None` | — |
| `list.remove(x)` | Xóa lần đầu tiên của `x` | ✅ | `None` | — |
| `list.clear()` | Xóa toàn bộ | ✅ | `None` | `ds = []` (object mới) |
