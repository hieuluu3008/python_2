# Brute-Force

**Brute-Force** (phương pháp vét cạn) là phương pháp giải quyết vấn đề bằng cách **thử tất cả các khả năng có thể** cho đến khi tìm ra đáp án đúng.

Không cần thông minh — chỉ cần thử hết.

## Cách Hoạt Động

Quy trình chung gồm 4 bước:

1. **Xác định không gian tìm kiếm** — tập hợp tất cả các lời giải có thể
2. **Liệt kê các ứng viên** — sinh ra từng lời giải tiềm năng
3. **Kiểm tra điều kiện** — xem ứng viên có thỏa mãn yêu cầu không
4. **Trả về kết quả** — dừng khi tìm thấy hoặc đã thử hết

```
for candidate in search_space:
    if is_valid(candidate):
        return candidate
```

## Ví Dụ 1 — Linear Search

Tìm kiếm tuyến tính: duyệt từng phần tử, so sánh với giá trị cần tìm.

````python
def linear_search(ds: list, target) -> int:
    """Tìm index của target trong ds, trả về -1 nếu không thấy."""
    for i, phan_tu in enumerate(ds):
        if phan_tu == target:
            return i
    return -1

so = [4, 7, 2, 9, 1, 5]
print(linear_search(so, 9))   # 3
print(linear_search(so, 99))  # -1
````

**Độ phức tạp:** O(n) — xấu nhất phải duyệt hết toàn bộ list.

## Ví Dụ 2 — Tìm Giá Trị Lớn Nhất

So sánh từng phần tử với giá trị lớn nhất hiện tại, cập nhật khi gặp phần tử lớn hơn.

````python
def tim_max(ds: list):
    """Tìm giá trị lớn nhất trong ds."""
    lon_nhat = ds[0]
    for phan_tu in ds[1:]:
        if phan_tu > lon_nhat:
            lon_nhat = phan_tu
    return lon_nhat

so = [3, 1, 4, 1, 5, 9, 2, 6]
print(tim_max(so))  # 9
````

## Ví Dụ 3 — Dò Mật Khẩu

Sinh tất cả tổ hợp 3 chữ số (000–999), thử từng cái cho đến khi khớp.

````python
def kiem_tra_matkhau(matkhau: str) -> bool:
    """Giả lập hàm kiểm tra mật khẩu."""
    return matkhau == "472"

def do_matkhau() -> str | None:
    """Thử tất cả mã 3 chữ số từ 000 đến 999."""
    for i in range(1000):
        thu = f"{i:03d}"          # đảm bảo luôn có 3 chữ số: 007, 042...
        if kiem_tra_matkhau(thu):
            return thu
    return None

ket_qua = do_matkhau()
print(f"Tìm thấy mật khẩu: {ket_qua}")  # Tìm thấy mật khẩu: 472
````

**Độ phức tạp:** O(10^n) với n là số chữ số — tăng theo hàm mũ.
- 3 chữ số → 1.000 lần thử
- 6 chữ số → 1.000.000 lần thử
- 10 chữ số → 10.000.000.000 lần thử

## Ưu & Nhược Điểm

| Ưu điểm | Nhược điểm |
|---|---|
| Đơn giản, dễ implement | Chậm với không gian lớn |
| Đảm bảo tìm ra đáp án nếu tồn tại | Độ phức tạp tăng nhanh (mũ / giai thừa) |
| Áp dụng được cho nhiều bài toán | Tốn tài nguyên (CPU, RAM) |

## Khi Nào Nên Dùng Brute-Force

✅ **Nên dùng khi:**
- Input nhỏ, tốc độ không quan trọng
- Cần baseline để kiểm tra tính đúng của thuật toán phức tạp hơn
- Chưa tìm ra hướng giải tốt hơn
- Bài toán yêu cầu kiểm tra toàn bộ (exhaustive search)

❌ **Không nên dùng khi:**
- Input lớn (n > 10^6 tùy bài)
- Không gian tìm kiếm tăng theo hàm mũ hoặc giai thừa
- Có thuật toán tốt hơn đã biết (binary search, dynamic programming...)

> **Quy tắc:** Luôn ước lượng độ phức tạp trước khi implement brute-force.

## Ước Lượng Độ Phức Tạp Nhanh

| Giới hạn n | Độ phức tạp chấp nhận được |
|---|---|
| n ≤ 10 | O(n!) — giai thừa |
| n ≤ 20 | O(2^n) — mũ |
| n ≤ 500 | O(n^3) |
| n ≤ 5.000 | O(n^2) |
| n ≤ 10^6 | O(n log n) |
| n ≤ 10^8 | O(n) |

Máy tính thông thường xử lý ~10^8 phép tính/giây.
