# Cấu Trúc Điều Kiện — Conditional Structure

Dùng để kiểm tra điều kiện và thực thi khối code tương ứng. Python dùng **indentation** (4 spaces) thay cho `{}` để xác định khối code.

## 1. if / elif / else

Cú pháp:
```
if <điều_kiện_1>:
    <khối code 1>
elif <điều_kiện_2>:
    <khối code 2>
else:
    <khối code mặc định>
```

- Chỉ **một nhánh** được thực thi — nhánh đầu tiên có điều kiện `True`
- `elif` và `else` là **tuỳ chọn**, có thể có nhiều `elif`
- Khi không có `else`, không có nhánh nào khớp → không làm gì cả

````python
diem = 75

if diem >= 90:
    xep_loai = "Xuất sắc"
elif diem >= 80:
    xep_loai = "Giỏi"
elif diem >= 65:
    xep_loai = "Khá"
elif diem >= 50:
    xep_loai = "Trung bình"
else:
    xep_loai = "Yếu"

print(f"Điểm {diem} → {xep_loai}")  # Điểm 75 → Khá
````

## 2. Toán Tử So Sánh

| Toán tử | Ý nghĩa | Ví dụ |
|---------|---------|-------|
| `==` | bằng giá trị | `5 == 5` → `True` |
| `!=` | khác giá trị | `5 != 3` → `True` |
| `>` | lớn hơn | `5 > 3` → `True` |
| `<` | nhỏ hơn | `5 < 3` → `False` |
| `>=` | lớn hơn hoặc bằng | `5 >= 5` → `True` |
| `<=` | nhỏ hơn hoặc bằng | `4 <= 5` → `True` |
| `is` | cùng object (identity) | `a is None` |
| `is not` | khác object | `a is not None` |
| `in` | nằm trong iterable | `2 in [1,2,3]` → `True` |
| `not in` | không nằm trong | `5 not in [1,2,3]` → `True` |

> **`==` vs `is`:** `==` so sánh **giá trị**, `is` so sánh **địa chỉ bộ nhớ**. Dùng `is` chỉ khi so sánh với `None`, `True`, `False`.

````python
# So sánh chuỗi dùng so sánh 
a = [1, 2, 3]
b = [1, 2, 3]
print(a == b)   # True  — cùng giá trị
print(a is b)   # False — khác object trong bộ nhớ

# Chaining so sánh — Python cho phép viết liền
x = 15
print(10 < x < 20)   # True — tương đương x > 10 and x < 20
print(1 <= x <= 100) # True
````

````python
if type(a) is list:
    print(a)
````

````python
ds = ["táo", "chuối", "cam"]
print("táo" in ds)       # True
print("xoài" not in ds)  # True

# Dùng 'in' với string
email = "user@gmail.com"
print("@" in email)      # True
````

## 3. Toán Tử Logic

| Toán tử | Ý nghĩa | Kết quả |
|---------|---------|----------|
| `and` | Cả hai `True` | `True and True` → `True` |
| `or` | Ít nhất một `True` | `False or True` → `True` |
| `not` | Phủ định | `not True` → `False` |

**Short-circuit evaluation:**
- `and` — dừng và trả về toán hạng đầu tiên có giá trị **Falsy**
- `or` — dừng và trả về toán hạng đầu tiên có giá trị **Truthy**

**Thứ tự ưu tiên:** `not` > `and` > `or`

````python
tuoi = 20
co_cccd = True

if tuoi >= 18 and co_cccd:
    print("Được phép đăng ký")

ngay_nghi = True
lam_them = False
if ngay_nghi or lam_them:
    print("Không đi làm hôm nay")
````

````python
# Short-circuit: 'and' trả về giá trị, không chỉ True/False
print(0 and "hello")    # 0       — dừng ở 0 (Falsy)
print(1 and "hello")    # 'hello' — qua hết, trả về cái cuối
print("" or "mặc định") # 'mặc định' — "" Falsy, tiếp tục qua 'or'
print("An" or "mặc định") # 'An'  — "An" Truthy, dừng lại
````

````python
# Ứng dụng thực tế: giá trị mặc định
ten_nhap = ""  # user nhập rỗng
ten_hien_thi = ten_nhap or "Khách"
print(f"Xin chào, {ten_hien_thi}")  # Xin chào, Khách
````

## 4. Giá Trị Truthy / Falsy

Python đánh giá mọi biểu thức như `bool` trong điều kiện.

**Falsy** (coi như `False`):
- `False`, `None`
- Số không: `0`, `0.0`, `0j`
- Chuỗi rỗng: `""`
- Collection rỗng: `[]`, `()`, `{}`, `set()`

**Truthy:** mọi thứ còn lại

````python
# Kiểm tra collection rỗng — không cần len()
ds = []
if not ds:
    print("List rỗng")   # in ra

du_lieu = {"ten": "An"}
if du_lieu:
    print("Dict có dữ liệu")  # in ra
````

````python
# Kiểm tra None — dùng 'is None', không dùng '== None'
ket_qua = None

if ket_qua is None:
    print("Chưa có kết quả")

# Tại sao không dùng 'not ket_qua'?
# Vì 0 và "" cũng Falsy — dễ nhầm lẫn khi giá trị hợp lệ là 0
ket_qua = 0
if not ket_qua:          print("Sai — 0 là kết quả hợp lệ nhưng bị coi là False")
if ket_qua is None:      print("Không in — vì 0 is not None")  # không in
if ket_qua is not None:  print("Đúng — 0 khác None")  # in ra
````

## 5. Ternary Expression (Biểu Thức Điều Kiện 1 Dòng)

Cú pháp: `<giá_trị_nếu_True> if <điều_kiện> else <giá_trị_nếu_False>`

Dùng để **gán biến** hoặc **return** đơn giản — tránh dùng cho logic phức tạp.

````python
x = 7
ket_qua = "chẵn" if x % 2 == 0 else "lẻ"
print(ket_qua)  # lẻ

# Dùng trong f-string
so_luong = 0
print(f"Có {so_luong} sản phẩm" if so_luong > 0 else "Hết hàng")  # Hết hàng

# Dùng trong list comprehension
so = [1, 2, 3, 4, 5, 6]
nhan = ["chẵn" if n % 2 == 0 else "lẻ" for n in so]
print(nhan)  # ['lẻ', 'chẵn', 'lẻ', 'chẵn', 'lẻ', 'chẵn']
````

## 6. match / case — Structural Pattern Matching (Python 3.10+)

Mạnh hơn `switch` trong các ngôn ngữ khác vì hỗ trợ **pattern matching** trên cấu trúc dữ liệu.

- `case _:` là wildcard, khớp mọi giá trị còn lại (tương đương `default`)
- Có thể khớp nhiều giá trị cùng lúc với `|`

````python
# match với giá trị đơn
ma_loi = 404

match ma_loi:
    case 200:
        print("OK")
    case 400:
        print("Bad Request")
    case 401 | 403:      # nhiều giá trị dùng '|'
        print("Không có quyền truy cập")
    case 404:
        print("Not Found")  # in ra
    case 500:
        print("Internal Server Error")
    case _:
        print("Mã lỗi không xác định")
````

````python
# match với tuple pattern (structural)
point = (0, 5)

match point:
    case (0, 0):
        print("Gốc tọa độ")
    case (0, y):              # y được bind giá trị 5
        print(f"Trên trục Y, y={y}")   # in ra
    case (x, 0):
        print(f"Trên trục X, x={x}")
    case (x, y):
        print(f"Điểm ({x}, {y})")
````

````python
point = (0, 0)

match point:
    case (x,y):
        print("z")
    case (0,0):
        print("a")
    case (0,y):
        print("b")
    case (x,y):
        print("c")
````

````python
# match với dict pattern
lenh = {"hanh_dong": "di_chuyen", "huong": "bac", "buoc": 3}

match lenh:
    case {"hanh_dong": "dung_lai"}:
        print("Dừng lại")
    case {"hanh_dong": "di_chuyen", "huong": huong, "buoc": buoc}:
        print(f"Di chuyển {huong} {buoc} bước")  # in ra
    case _:
        print("Lệnh không hợp lệ")
````

## 7. Điều Kiện Lồng Nhau & Early Return

Tránh lồng quá sâu (> 2 cấp). Kỹ thuật **early return / early exit** giúp code phẳng hơn, dễ đọc hơn.

````python
# Lồng nhau sâu — khó đọc
def xu_ly_don_hang_v1(don):
    if don is not None:
        if don["so_luong"] > 0:
            if don["da_thanh_toan"]:
                return f"Xử lý đơn: {don['ma_don']}"
            else:
                return "Chưa thanh toán"
        else:
            return "Số lượng không hợp lệ"
    else:
        return "Đơn hàng rỗng"

# Early return — flat, dễ đọc
def xu_ly_don_hang_v2(don):
    if don is None:
        return "Đơn hàng rỗng"
    if don["so_luong"] <= 0:
        return "Số lượng không hợp lệ"
    if not don["da_thanh_toan"]:
        return "Chưa thanh toán"
    return f"Xử lý đơn: {don['ma_don']}"

don = {"ma_don": "ORD-001", "so_luong": 2, "da_thanh_toan": False}
print(xu_ly_don_hang_v2(don))   # Xử lý đơn: ORD-001
print(xu_ly_don_hang_v2(None))  # Đơn hàng rỗng
````

## 8. Một Số Pattern Thực Tế

Các cách viết điều kiện phổ biến trong Python thực tế.

````python
# Pattern 1: Gán giá trị mặc định với 'or'
config = {}
timeout = config.get("timeout") or 30
print(timeout)  # 30

# Pattern 2: Walrus operator ':=' — gán và kiểm tra cùng lúc (Python 3.8+)
import re #regular expression: search, match, split, ....
text = "Email: user@example.com"
if m := re.search(r"[\w.]+@[\w.]+", text): # m là match object
    print(f"Tìm thấy email: {m.group()}")  # user@example.com
## Tương đương:
# m = re.search(r"[\w.]+@[\w.]+", text)  # gọi hàm
# if m:                                   # kiểm tra
#     print(f"Tìm thấy email: {m.group()}")

# Pattern 3: any() / all() thay cho chuỗi or/and
quyen = ["read", "write", "delete"]
can_quyen = ["read", "admin"]

if any(q in quyen for q in can_quyen):
    print("Có ít nhất 1 quyền phù hợp")  # in ra

if all(q in quyen for q in ["read", "write"]):
    print("Có đủ tất cả quyền cần thiết")  # in ra
````

````python
# Pattern 4: Dict thay cho chuỗi if/elif (dispatch table)
def cong(a, b): return a + b
def tru(a, b):  return a - b
def nhan(a, b): return a * b
def chia(a, b): return a / b if b != 0 else "Lỗi chia cho 0"

phep_tinh = {"+": cong, "-": tru, "*": nhan, "/": chia}

toan_tu = "*"
ham = phep_tinh.get(toan_tu)
if ham:
    print(ham(10, 3))  # 13
````

````python
class May_tinh:
    def __init__(self):
        self.phep_tinh = {
            "+": self.cong,
            "-": self.tru,
            "*": self.nhan,
            "/": self.chia,
        }

    def cong(self,a, b): return a + b
    def tru(self,a, b):  return a - b
    def nhan(self,a, b): return a * b
    def chia(self,a, b): return a / b if b != 0 else "Lỗi chia cho 0"

    def tinh(self,toan_tu,a,b):
        ham = self.phep_tinh.get(toan_tu)
        if ham:
            return ham(a,b)
        else:
            return f"Không hỗ trợ"

mt = May_tinh()
mt.tinh("*",2,3)
````

## Tóm Tắt

| Cú pháp | Dùng khi |
|---------|----------|
| `if/elif/else` | Logic phân nhánh tổng quát |
| Ternary `A if C else B` | Gán biến đơn giản 1 dòng |
| `match/case` | So sánh một giá trị với nhiều pattern (Python 3.10+) |
| `any()` / `all()` | Kiểm tra điều kiện trên nhiều phần tử |
| Dict dispatch | Thay chuỗi `elif` khi map giá trị → hành động |
| Early return | Giảm lồng nhau, làm phẳng logic |

**Nguyên tắc chung:** Ưu tiên code phẳng hơn code lồng nhau. Nếu `elif` quá nhiều → cân nhắc `match/case` hoặc dict dispatch.

### Luyện Tập

- **Bài 1:** Viết hàm `xep_loai_bmi(bmi)` trả về: `"Thiếu cân"` (< 18.5), `"Bình thường"` (18.5–24.9), `"Thừa cân"` (25–29.9), `"Béo phì"` (≥ 30). Dùng chaining so sánh.
- **Bài 2:** Cho list `san_pham = [{"ten": "A", "gia": 100, "ton_kho": 0}, {"ten": "B", "gia": 50, "ton_kho": 5}]`. Dùng ternary trong list comprehension tạo list `["Còn hàng", "Hết hàng"]` tương ứng.
- **Bài 3:** Dùng `match/case` viết hàm `ten_ngay(so)` nhận 1–7, trả về tên ngày. Ngoài khoảng → `"Không hợp lệ"`.
- **Bài 4:** Viết hàm `kiem_tra_dang_nhap(user)` nhận dict có `email`, `password`, `is_active`. Dùng early return: thiếu email → lỗi, sai password → lỗi, tài khoản bị khoá → lỗi, còn lại → thành công.

````python
# Bài 1
def xep_loai_bmi(bmi):
    if bmi < 18.5:
        return "Thieu can"
    elif bmi < 24.9:
        return "Binh thuong"
    elif bmi < 29.9:
        return "Thua can"
    elif bmi > 30:
        return "Beo phi"

xep_loai_bmi(25)
````

````python
# Bài 2
san_pham = [{"ten": "A", "gia": 100, "ton_kho": 0}, {"ten": "B", "gia": 50, "ton_kho": 5}]
# a = [A if True else B for loop] # if ternary / Nếu không có else thì [A for loop if True]
a = ["Còn hàng" if san_pham[i]["ton_kho"] > 0 else "Hết hàng" for i in range(len(san_pham))]
print(a)
````

````python
# Bài 3
def ten_ngay(so):
    match so:
        case 1:
            return "Ngay 1"
        case 2:
            return "Ngay 2"
        case _:
            return "Khong hop le"

ten_ngay(1)
````

````python
# Bài 4
def kiem_tra_dang_nhap(user):
    password = [123, 456, 789]
    if user.get("email") is None:
        return f"Khong co email"
    if user.get("password") not in password:
        return f"Sai mat khau"
    if user.get("is_active") is False:
        return f"Tai khoan bi khoa"
    return f"Dang nhap thanh cong"

kiem_tra_dang_nhap({"email":"abc","password":123,"is_active":False})
````
